# 柏の気温を調べる：朝と昼、どれくらい違う？

**氏名：［永井港人］**

## 調べたいこと
柏付近の2026年10月15日について、朝8時と昼14時の気温は何℃違うかを調べる。保存されたJSONデータをPythonの辞書・リストで読み取り、数値を比較する。


## データと出典
- データ：Open-Meteoの時間別気温（`temperature_2m`）
- 出典URL：https://api.open-meteo.com/v1/forecast?latitude=35.85&longitude=140.00&hourly=temperature_2m&start_date=2026-10-15&end_date=2026-10-16&timezone=Asia%2FTokyo
- 取得日：2026年10月15日（配布された保存JSONの記録に基づく）
- 使用ファイル：`weather-kashiwa-example.json`（このNotebookと同じフォルダに置く）

今回はAPIへ再接続せず、配布されたJSONを使うため、通信できない環境でも実行できる。


```python
import json
from pathlib import Path

json_path = Path("weather-kashiwa-example.json")

if not json_path.exists():
    json_path = next(
        Path.cwd().parent.rglob("weather-kashiwa-example.json")
    )

with open(json_path, encoding="utf-8") as file:
    data = json.load(file)

times = data["hourly"]["time"]
temperatures = data["hourly"]["temperature_2m"]

print("読み込み成功！")
print("時刻データの件数:", len(times))
print("気温データの件数:", len(temperatures))
```

## 朝8時と昼14時の気温を比べる
`hourly`は辞書、`time`と`temperature_2m`はリストである。時刻の文字列を検索して、2026年10月15日の8時と14時に当たる位置を探す。


```python
target_date = "2026-10-15"
morning_time = target_date + "T08:00"
afternoon_time = target_date + "T14:00"

morning_index = times.index(morning_time)
afternoon_index = times.index(afternoon_time)

morning = temperatures[morning_index]
afternoon = temperatures[afternoon_index]
difference = afternoon - morning

print(f"対象日: {target_date}")
print(f"朝8時: {morning} ℃")
print(f"昼14時: {afternoon} ℃")
print(f"昼14時 − 朝8時: {difference:.1f} ℃")
```

## 1日の気温の変化も確認する
リストの`max`と`min`を使い、10月15日の最高値・最低値と、その差を調べる。ここでいう最高・最低は、このJSONに含まれる時間別データの範囲内での値である。


```python
day_indices = [i for i, time in enumerate(times) if time.startswith(target_date)]
day_temperatures = [temperatures[i] for i in day_indices]

max_temp = max(day_temperatures)
min_temp = min(day_temperatures)
max_index = day_indices[day_temperatures.index(max_temp)]
min_index = day_indices[day_temperatures.index(min_temp)]

print(f"時間別データの最高気温: {max_temp} ℃（{times[max_index][11:]}）")
print(f"時間別データの最低気温: {min_temp} ℃（{times[min_index][11:]}）")
print(f"最高と最低の差: {max_temp - min_temp:.1f} ℃")
```

## 分かったこと・確認・限界

- **分かったこと：** 2026年10月15日は、朝8時が17.7℃、昼14時が21.1℃で、昼14時の方が3.4℃高かった。時間別データの最高は21.3℃（15:00）、最低は16℃（06:00）で、差は5.3℃だった。
- **元データとの確認：** `weather-kashiwa-example.json`の`hourly.time`と`hourly.temperature_2m`を照合し、朝8時の値が17.7、昼14時の値が21.1であることを確認した。
- **このデータだけでは分からないこと：** 気温が変化した理由や、実際に観測された気温かどうかは、このJSONだけでは確定できない。また、別の日や柏市内の別の場所にも同じ傾向があるとは限らない。
- **出典URL：** https://open-meteo.com/en/docs （Open-Meteo API）／データ取得日：2026年10月15日（配布JSONの記録に基づく）。
