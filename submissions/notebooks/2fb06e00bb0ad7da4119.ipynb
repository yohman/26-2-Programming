# ゲーム相棒ピクセル｜冒険の作戦会議
ゲームが好きな人のために、勝ったときは一緒に喜び、疲れたときは休憩をすすめる相棒Botです。

このBotはAIではなく、**自分で決めたルール**で返します。

## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# 辞書（dict）：Botのプロフィール
bot = {
    "name": "ピクセル",
    "role": "ゲームの冒険相棒",
    "likes": ["RPG", "レベル上げ", "ポーション"],
}

# 気分ごとの顔
faces = {
    "normal": "🎮",
    "happy": "🎮✨",
    "tired": "🎮💤",
}

# プロフィールを表示する
print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
print(f"好きなもの：{', '.join(bot['likes'])}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random

# 気分ごとの返事リスト
responses = {
    "normal": [
        f"やあ！{bot['name']}だよ。次の冒険の作戦を考えよう！",
        "今どんなゲームをしているの？",
        f"ぼくは{bot['likes'][0]}が好き！おすすめのゲームはある？",
        "焦らず一歩ずつ進めば、きっと道が見つかるよ。",
    ],
    "happy": [
        "ナイスプレイ！その調子で冒険を続けよう！",
        "やったね！勝利のポーズ！🎉",
        "その勢い、まるで伝説の勇者だね！",
        f"次は{bot['likes'][1]}で、もっと強くなろう！",
    ],
    "tired": [
        "おつかれ！セーブして少し休憩しよう。",
        "負けても大丈夫。次の作戦を考えればいいよ！",
        "水分補給も忘れずに。休憩も大事な作戦だよ。",
        f"{bot['likes'][2]}を使うみたいに、気分を回復しよう！",
    ],
}

print("気分ごとの返事を準備したよ！")
print("元気なときの例：" + responses["happy"][0])
```

## 3｜気分チェッカー
一回の入力に一回返事をします。ゲームの話題や気分を表す言葉に反応します。
「クリアした！」「負けた」「ボス」「こんにちは」などを試してください。

**オリジナル機能：**「ボス」という言葉が入ると、気分に関係なくボス戦用の作戦メッセージを返します。

```python
# 気分を判定する言葉のリスト
happy_words = [
    "元気", "楽しい", "うれしい", "嬉しい", "最高",
    "勝った", "クリア", "レベルアップ", "神プレイ",
    "happy", "good", "great", "fun",
]
tired_words = [
    "元気じゃない", "つかれ", "疲れ", "眠い", "しんど",
    "負けた", "詰んだ", "難しい", "もう無理",
    "tired", "sleepy", "sad", "bad",
]

# Botが質問する
question = "今日のゲームはどう？ 勝った？ 疲れた？"
print(f"{faces['normal']} {bot['name']}：{question}")

# 入力を一回受け取る
try:
    message = input("あなた：").strip().lower()
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 気分を決める
mood = "normal"

# 先に疲れた気分を調べ、あとから元気な気分を上書きしないようにする
for word in tired_words:
    if word in message:
        mood = "tired"
        break

# 「元気じゃない」などを誤判定しないよう、tiredでなければhappyを調べる
if mood != "tired":
    for word in happy_words:
        if word in message:
            mood = "happy"
            break

# オリジナル機能：「ボス」に反応する特別メッセージ
if "ボス" in message:
    reply = "ボス戦だね！相手の動きをよく見て、回復アイテムも準備しよう！"
else:
    reply = random.choice(responses[mood])

print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **工夫したところ：** 「元気じゃない」のように、元気を表す言葉と疲れた気分の言葉が重なる場合があります。疲れた気分を先に判定し、誤判定を減らしています。

## 4｜自分のキャラクターにする
- Botの名前は「ピクセル」、役割は「ゲームの冒険相棒」です。
- RPGやレベル上げなど、ゲームに関係する返事を用意しました。
- 「勝った」「クリア」「負けた」などの言葉から気分を判定します。
- 「ボス」という言葉には、特別な作戦メッセージを返します。

「勝った」「クリア」「負けた」などの言葉から気分を判定し、「ボス」と入力するとボス戦の作戦メッセージを返す機能を追加しました。ゲーム好きの人が楽しめるように、返事もゲームらしい言葉にしました。
