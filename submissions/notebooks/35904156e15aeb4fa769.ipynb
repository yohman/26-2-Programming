## 1. 問い・出典・対象データの説明

**【問い】** 千葉県内で発生している自転車盗難の実態はどのようなものか？また、どの年齢層やどのような状況で被害が多いのかを明らかにし、有効な防犯対策を提案する。
**【出典】** 千葉県オープンデータサイト（自転車盗難に関するオープンデータ）
**【対象データ】** `bike.csv` は、千葉県内で発生した自転車盗難の被害報告データをまとめたものです。**データ内の1行は「1件の自転車盗難被害の報告」**を表しており、発生場所、被害者の年齢層、施錠の有無などが記録されています。

```python
import pandas as pd

# データの読み込み
df = pd.read_csv("bike.csv", encoding="cp932")

# 1. for文による集計（施錠関係の件数）
lock_counts = {"施錠した": 0, "施錠せず": 0}
for status in df["施錠関係"]:
    if status in lock_counts:
        lock_counts[status] += 1

print("【施錠関係の集計】")
for k, v in lock_counts.items():
    print(f"{k}: {v}件")


# 2. ランキング（発生場所ワースト3）
print("\n【発生場所トップ3ランキング】")
place_ranking = df["発生場所"].value_counts().head(3)

rank = 1
for place, count in place_ranking.items():
    print(f"第{rank}位: {place} ({count}件)")
    rank += 1


# 3. 図（被害者の年齢トップ5の自転車グラフ）
# ※ 🚲1つを約500件として可視化
print("\n【被害者の年齢層トップ5（自転車グラフ）】")
age_ranking = df["被害者の年齢"].value_counts().head(5)
total = len(df)

for i, (age, count) in enumerate(age_ranking.items()):
    bar = "🚲" * (count // 500)
    percent = (count / total) * 100
    alert = " 🚨ダントツ！" if i == 0 else ""
    print(f"{age:4}: {bar:<9} ({count:4}件 / {percent:4.1f}%){alert}")
```

## 分析からの発見と防犯の提案

### 数値を使った発見
1.全体の被害件数のうち、「施錠せず」の状態で被害に遭ったケースが6,879件に上り、「施錠した」状態（3,175件）の約2倍以上となっています。
無施錠によって盗難リスクを高めるかが明確に表れています。
2. 文字の棒グラフからも分かる通り、被害者の年齢層は10歳代（4,460件）が突出して多く、次いで20歳代（2,615件）となっています。
この2つの年代だけで全体の約7割を占めており、若年層の被害が顕著です。

### 防犯の提案
「若年層に向けた『短時間でも施錠』の徹底とツーロックの推進」

データから、10代〜20代の若年層が、無施錠の状態で被害に遭うケースが多いことが推測されます。
そのため、中学校や高校、大学と連携し、「駐輪場や自宅に停める際、わずかな時間でも必ず鍵をかける」という基本の啓発活動を強化すべきです。
あわせて、備え付けの鍵に加えてワイヤーロック等を用いる「ツーロック」を推進することで、盗難リスクを大幅に下げることができます。

## 5. データ照合による確認と限界、for文の役割

**【データの確認と限界】**
施錠関係の集計結果（施錠した: 3175件、施錠せず: 6879件）の合計は10,054件となり、全体のデータ件数と一致することを確認しました。ただし、このデータは「警察に届け出があった被害」のみを集計したものであり、泣き寝入りなどで報告されていない潜在的な被害は含まれていないという限界があります。

**【for文の役割】**
最初の集計では、データの行数（10,054件）だけ `for` 文を繰り返し回しています。1回のループごとに1件の「施錠関係」のデータを取り出し、それが「施錠した」か「施錠せず」かを判定して、辞書（`lock_counts`）の該当するカウントを+1していく役割を果たしています。
