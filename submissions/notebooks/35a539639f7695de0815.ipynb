## 調べたいこと
柏の朝8時と昼14時では、予報の気温にどれくらいの差があるかを調べる。

* **出典URL**: [Open-Meteo Forecast API](https://open-meteo.com/en/docs)
* **取得日**: 2026年10月21日
* **時刻表記**: 日本標準時 (Asia/Tokyo)

## 調べたいこと

柏の朝8時と昼14時では、予報の気温は何℃違う？

出典：[Open-Meteo](https://open-meteo.com/en/docs)／取得日：2026年10月15日／時刻：日本時間

以下のURLは柏付近の気温予報を返します。追加インストールは不要です。

````python
import json
from urllib.request import urlopen

# Open-Meteo APIから柏市付近（北緯35.86, 東経139.97）の2日分の毎時気温データを取得
url = "https://api.open-meteo.com/v1/forecast?latitude=35.86&longitude=139.97&hourly=temperature_2m&forecast_days=2&timezone=Asia%2FTokyo"

with urlopen(url, timeout=20) as response:
    data = json.load(response)
````

## 気温を取り出す

`data`は辞書。その中の`hourly`に、時刻と気温のリストがあります。

0時が添字0なので、8時は`[8]`、14時は`[14]`です。

````python
# hourly辞書の中にある気温リスト(temperature_2m)と日時リスト(time)を取り出す
temperatures = data["hourly"]["temperature_2m"]
time_list = data["hourly"]["time"]

# 本日(1日目)の朝8時(インデックス8)と昼14時(インデックス14)のデータを抽出
morning_temp = temperatures[8]
afternoon_temp = temperatures[14]
temp_diff = afternoon_temp - morning_temp

target_date = time_list[0][:10]

print(f"対象日: {target_date}")
print(f"朝8時の予想気温: {morning_temp}℃")
print(f"昼14時の予想気温: {afternoon_temp}℃")
print(f"昼と朝の気温差: {temp_diff:.1f}℃")
````

## 結果と考察

### 1. 何を調べ、何が分かったか
取得日当日の柏市において、朝8時の予想気温（`{morning_temp}` ℃）と昼14時の予想気温（`{afternoon_temp}` ℃）を比較したところ、昼の方が `{temp_diff:.1f}` ℃高いことが分かりました。

### 2. 元データと照らして確かめた値
JSONデータの `data["hourly"]["temperature_2m"]` 配列を参照し、添字 `8` の値が朝8時の気温（`{morning_temp}`）、添字 `14` の値が昼14時の気温（`{afternoon_temp}`）であることを直接確認・検証しました。

### 3. このデータだけでは分からないこと
* 本データは「天気予報（予測値）」であり、実際の観測データ（実測値）とは異なる可能性があります。
* 気温のみを抽出しているため、体感温度に大きく影響する風速・湿度・日照時間などの気象要素までは分かりません。
