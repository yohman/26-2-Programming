## 2｜CSVを読む

`latitude`は緯度、`longitude`は経度、`mag`はマグニチュード。CSVの値は文字なので、地図に使う緯度・経度は`float()`で数に変えます。


```python
import csv
import folium

csv_file = "csv_file.csv"

with open(csv_file, encoding="utf-8-sig") as file:
    earthquakes = list(csv.DictReader(file))

print(f"{len(earthquakes)}件のデータを読みました")
```

## 3｜一行につき、マーカーを一つ

実行する前に、どこにどんな点が出るか予想しよう。`for`の中が各地震に対して繰り返されます。最後の`earthquake_map`でNotebookに地図を表示します。


```python
# 1. 地図の土台を作る（初期位置を日本近海にする）
earthquake_map_v2 = folium.Map(location=[35.0, 137.0], zoom_start=4,)

# 2. データを1件ずつループ処理
for quake in earthquakes:
    
    # 【エラー対策】マグニチュードや緯度経度が空欄のデータはスキップする
    if not quake["mag"] or not quake["latitude"] or not quake["longitude"]:
        continue
        
    # 地震(earthquake)以外のデータ（mining explosionなど）は除外する
    if quake["type"] != "earthquake":
        continue

    # 文字列を数値（小数）に変換
    lat = float(quake["latitude"])
    lon = float(quake["longitude"])
    mag = float(quake["mag"])

    # 【改良1】if文を使い、マグニチュードの大きさによって「色」と「円の大きさ（半径）」を自動で変える
    if mag >= 5.0:
        dot_color = "red"          # M5以上：赤色（大地震）
        dot_radius = mag * 3       # 円を大きく
    elif mag >= 4.0:
        dot_color = "orange"       # M4以上：オレンジ色（中地震）
        dot_radius = mag * 2
    else:
        dot_color = "gold"         # M4未満：黄色（小地震）
        dot_radius = mag * 1.5     # 円を小さく

    # 【改良2】クリックした時に表示する詳細情報（ポップアップ）を整理して追加
    # HTMLの<br>タグで改行し、<strong>タグで文字を太字にしています
    popup_text = f"""
    <strong>M {mag} 地震情報</strong><br>
    場所: {quake['place']}<br>
    日時: {quake['time']}<br>
    深さ: {quake['depth']} km
    """

    # マウスを重ねたときに出る簡易ツールチップ
    tooltip_text = f"M {mag}"

    # 【改良3】folium.Markerを「folium.CircleMarker」に変更し、半透明の円で地図に描く
    folium.CircleMarker(
        location=[lat, lon],
        radius=dot_radius,         # 改良1で決まった半径
        color=dot_color,           # 改良1で決まった枠線の色
        fill=True,
        fill_color=dot_color,      # 改良1で決まった塗りつぶしの色
        fill_opacity=0.5,          # 重なりが見やすくなるように半透明(50%)にする
        popup=folium.Popup(popup_text, max_width=250), # 改良2のポップアップ
        tooltip=tooltip_text
    ).add_to(earthquake_map_v2)

# 改良された地図を表示
earthquake_map_v2
```
