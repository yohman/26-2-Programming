# 柏の朝から昼にかけての気温の変化

**氏名：［今井恭瑳］**

```python
import json
from urllib.request import urlopen

url = "https://api.open-meteo.com/v1/forecast?latitude=35.86&longitude=139.97&hourly=temperature_2m&forecast_days=2&timezone=Asia%2FTokyo"

with urlopen(url, timeout=20) as response:
    data = json.load(response)
```

```python
temperatures = data["hourly"]["temperature_2m"]
morning = temperatures[7]
morning1 = temperatures[8]
morning2 = temperatures[9]
morning3 = temperatures[10]
afternoon = temperatures[11]
afternoon1 = temperatures[12]
afternoon2 = temperatures[13]
afternoon3 = temperatures[14]
afternoon4 = temperatures[15]

print(f"対象日: {data['hourly']['time'][0][:10]}")
print(f"朝7時: {morning} ℃")
print(f"朝8時: {morning1} ℃")
print(f"朝9時: {morning2} ℃")
print(f"朝10時: {morning3} ℃")
print(f"昼11時: {afternoon} ℃")
print(f"昼12時: {afternoon1} ℃")
print(f"昼13時: {afternoon2} ℃")
print(f"昼14時: {afternoon3} ℃")
print(f"昼15時: {afternoon4} ℃")
```

＃発見
　朝7時から昼の11時にかけて気温が上昇していき、昼の13時の気温が最も高くそれ以降は気温は下がってゆく。
＃確かめた値
　10月21日の柏市朝7時から昼15時までの気温
＃限界
　雨や風などは考慮されていない
