# ソラのゲーム開発相棒Bot
ゲーム制作を頑張る人のために、気分に合わせて励ましたり、次の一歩を提案したりするBotです。

このBotはAIではなく、辞書・リスト・条件分岐で決めたルールに従って返事をします。


## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# 辞書（dict）でBotのプロフィールを設定する
bot = {
    "name": "ソラ",
    "role": "ゲーム開発の相棒",
    "likes": ["Unity", "オリジナルゲーム作り", "新しいアイデア"],
}

# 気分ごとの顔を辞書で設定する
faces = {
    "normal": "🎮",
    "happy": "🎮✨",
    "tired": "🎮💤",
}

print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
print("好きなもの：" + "、".join(bot["likes"]))
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random

# 気分ごとの返事をリストで用意する
responses = {
    "normal": [
        f"こんにちは！{bot['name']}だよ。今日はゲーム制作の何を進める？",
        "小さく一歩進めるだけでも、立派な前進だよ。",
        f"私は{bot['likes'][0]}が好き！今日は何を作っているの？",
        "アイデアをメモしておくと、あとで役立つよ。",
    ],
    "happy": [
        "いいね！その勢いで一つ完成させよう！",
        "最高！動くところまで作れると達成感があるよね。",
        "そのアイデア、もっと聞かせて！",
        f"私も{bot['likes'][1]}を応援しているよ！",
    ],
    "tired": [
        "おつかれさま！少し休憩してから再開しよう。",
        "全部を一気にやらなくて大丈夫。一つずつ進めよう。",
        "エラーは上達のチャンス。原因を一つずつ確認しよう。",
        f"{bot['likes'][2]}をメモして、今日は休むのもありだよ。",
    ],
}

print("元気なときの返事の例：" + responses["happy"][0])
print("疲れたときの返事の例：" + responses["tired"][0])
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 気分を判定するキーワードのリスト
happy_words = [
    "元気", "楽しい", "うれしい", "嬉しい", "最高",
    "できた", "完成", "順調", "happy", "great",
]
tired_words = [
    "元気じゃない", "疲れ", "つかれ", "眠い", "しんどい",
    "つらい", "エラー", "できない", "困った", "tired", "sad",
]

# 反応する言葉を追加することもできます
game_words = ["unity", "ゲーム", "開発", "プログラム", "コード"]

question = "今の気分や、ゲーム制作で起きたことを教えて！"

# Notebookを全Cell実行したときに出力が残るよう、サンプル入力を設定
# 別の言葉を試したいときは、この文字列を変更して再実行してください。
message = "Unityでゲーム作りが楽しい！"

print(f"{faces['normal']} {bot['name']}：{question}")
print(f"👤 あなた：{message}")

# 条件分岐で気分を決める。初期状態はふつう。
mood = "normal"

# 先に疲れた気分を確認することで「元気じゃない」も正しく判定する
for word in tired_words:
    if word in message.lower():
        mood = "tired"
        break
else:
    for word in happy_words:
        if word in message.lower():
            mood = "happy"
            break

# 返事を一つ選ぶ
reply = random.choice(responses[mood])

# ゲーム制作に関する言葉が含まれていれば、相棒らしい一言を加える
if any(word in message.lower() for word in game_words):
    reply += " ゲーム制作は小さな積み重ね！"

print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **工夫した点：** 「元気じゃない」のように、元気を表す言葉と疲れを表す言葉が重なる場合があります。先に疲れた気分を判定することで、誤判定を減らしています。
