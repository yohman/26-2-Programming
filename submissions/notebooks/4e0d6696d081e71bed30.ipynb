## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
bot = {
    "name": "もちものくん",
    "role": "忘れ物がないかチェックする相棒",
    "likes": ["持ち物チェック", "整理整頓", "お出かけ"],
}
faces = {
    "normal": "🤖",
    "happy": "🤖✨",
    "tired": "🤖💤",
}
print(f"{faces['normal']} {bot['name']}：{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random

responses = {
    "normal": [
        "スマホ、財布、鍵は持った？",
        "学生証を忘れていないか確認しよう！",
        "今日の授業で使う教科書とノートは入れた？",
        "パソコンと充電器も確認してね！",
    ],
    "happy": [
        "準備ばっちり！最後にスマホ、財布、鍵を確認！",
        "いい調子！提出する課題も忘れずにね！",
        "お出かけ前に、かばんの中をもう一度チェック！",
        "持ち物がそろったら、時間に余裕を持って出発しよう！",
    ],
    "tired": [
        "疲れているときは、一つずつ持ち物を確認しよう。",
        "まずはスマホ、財布、鍵の三つを確認してね。",
        "慌てずに、学生証が入っているか見てみよう。",
        "明日の持ち物を今のうちに準備すると、朝が楽になるよ。",
    ],
}
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜忘れ物チェッカー
1回の実行で1回返事をします。続けるときは、このセルをもう一度実行します。
「確認」「準備できた」「疲れた」「終了」を試せます。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。


```python
# 1. 入力を判定する言葉のリスト
happy_words = [
    "準備できた",
    "準備完了",
    "ばっちり",
    "持った",
]

tired_words = [
    "準備できてない",
    "持ってない",
    "疲れた",
    "つかれた",
    "眠い",
]

# 2. Botが質問する
question = "持ち物を確認しよう！「確認」「準備できた」「疲れた」などを入力してね。「終了」で終わるよ。"
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える
try:
    message = input("あなた：").strip().lower()
except KeyboardInterrupt:
    message = "終了"

print(f"あなた：{message}")

# 4. 返事の種類を決める
mood = "normal"
matched = False

for word in happy_words:
    if word in message:
        mood = "happy"
        matched = True

# 後に調べるので、両方に当てはまったら tired を優先する
for word in tired_words:
    if word in message:
        mood = "tired"
        matched = True

# 5. 入力に合わせて返す
if message == "終了":
    print(f"{faces['happy']} {bot['name']}：いってらっしゃい！気をつけてね！")

elif "確認" in message or matched:
    reply = random.choice(responses[mood])
    print(f"{faces[mood]} {bot['name']}：{reply}")

else:
    print(f"{faces['normal']} {bot['name']}：その言葉にはまだ対応していないよ。")
    print("「確認」「準備できた」「疲れた」「終了」などを入力してね。")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
