```python
import csv
```

```python
# CSVファイルの読み込み（ダウンロードフォルダから直接読み込み）
data = []
with open(r'[local]\Downloads\japan (1).csv', mode='r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)  # ヘッダー（列名）を取得
    for row in reader:
        data.append(row)

# データの確認
print("--- ヘッダー（列名） ---")
print(header[:5])

print("\n--- データの先頭5行 ---")
for r in data[:5]:
    print(r[:5])

print(f"\n総データ件数: {len(data)} 件")
```

```python
# 65歳以上人口が100万人（1000000人）を超える都道府県をforループでカウント・抽出
high_elderly_count = 0
high_elderly_list = []

for row in data:
    pref = row[0]              # 都道府県名
    elderly_pop = int(row[4])  # 65歳以上人口（整数に変換）
    
    if elderly_pop >= 1000000:
        high_elderly_count += 1
        high_elderly_list.append((pref, elderly_pop))

print(f"65歳以上人口が100万人以上の都道府県数: {high_elderly_count} 地域\n")
print("【対象の都道府県一覧】")
for pref, pop in high_elderly_list:
    print(f"- {pref}: {pop:,} 人")
```

```python
# 結果をテキスト一覧として表示
print("=== 65歳以上人口 100万人以上の地域一覧 ===")
for pref, pop in high_elderly_list:
    print(f"都道府県: {pref:<6} | 65歳以上人口: {pop:,} 人")
```

2. forループとif条件判定文を組み合わせることで、65歳以上人口が100万人を超える主要な都道府県を正確に抽出・カウントできた。
3. 抽出結果から、首都圏や近畿圏などの大都市圏を擁する都道府県において、高齢者人口が著しく多い傾向があることが確認できた。
