## 自転車盗難の発生場所から考える身近な防犯対策

氏名：井坂美晴

## 調査の問い
自転車盗難はどのような場所で多く発生しているのか。また、発生場所と施錠状況の関係から、どのような防犯対策が考えられるか。

## 調査の目的
自転車盗難の発生場所を比較し、どのような場所で被害が多く発生しているのかを調べる。また、場所ごとの施錠状況も確認し、分析結果をもとに身近にできる防犯対策について考える。

## 使用するデータ

今回の調査では、千葉県警察が公開している「令和7年中の発生情報」の自転車盗データを使用する。
今回は特に「発生場所」と「施錠関係」の項目に注目する。発生場所ごとの件数を比較し、さらに場所によって施錠状況に違いがあるのかを調べる。

データの出典：
千葉県警察 オープンデータ
https://www.police.pref.chiba.jp/seisoka/safe-life_publicspace-statistics_00002.html

```python
filename = "bike.csv"
encoding = "cp932"

import csv

with open(filename, encoding=encoding, newline="") as file:
    incidents = list(csv.DictReader(file))

print("データ件数：", len(incidents))
print("項目名：", list(incidents[0].keys()))
```

## 発生場所ごとの集計

自転車盗難がどのような場所で発生しているのかを調べる。

```python
# 発生場所ごとの件数を集計する
location_counts = {}

for row in incidents:
    location = row["発生場所"]

    if location == "":
        location = "不明・未記入"

    if location in location_counts:
        location_counts[location] += 1
    else:
        location_counts[location] = 1

# 結果を表示する
for location, count in location_counts.items():
    print(location, ":", count, "件")
```

## 発生場所のランキング

次に、発生場所ごとの件数を多い順に並べる。

```python
# 件数の多い順に並べ替える
ranking = sorted(
    location_counts.items(),
    key=lambda x: x[1],
    reverse=True
)

# ランキングを表示する
for i, (location, count) in enumerate(ranking, start=1):
    print(f"{i}位：{location}（{count}件）")
```

## 発生場所ごとの割合

件数だけでなく、全体に対してそれぞれの場所がどのくらいの割合を占めているのかも調べる。
割合を見ることで、特定の場所に盗難の記録が集中しているかを確認する。

```python
total = len(incidents)

for location, count in ranking:
    percentage = count / total * 100
    print(f"{location}：{count}件（{percentage:.1f}%）")
```

## 発生場所ごとのグラフ

集計した結果を棒グラフにして、発生場所ごとの件数を視覚的に比較する。

```python
# 発生場所ごとの件数を横棒グラフで表示する

max_count = max(count for location, count in ranking)

for location, count in ranking:
    bar = "■" * int(count / max_count * 30)
    print(f"{location:15} | {bar} {count}件")
```

## 発生場所と施錠状況の関係

次に、発生場所によって施錠状況に違いがあるのかを調べる。
自転車盗難の記録には、施錠していた場合と施錠していなかった場合がある。場所ごとにそれぞれの件数を集計し、どのような傾向があるのかを比較する。

```python
# 発生場所ごとに施錠状況を集計する
lock_counts = {}

for row in incidents:
    location = row["発生場所"]
    lock = row["施錠関係"]

    if location == "":
        location = "不明・未記入"

    if lock == "":
        lock = "不明・未記入"

    if location not in lock_counts:
        lock_counts[location] = {}

    if lock in lock_counts[location]:
        lock_counts[location][lock] += 1
    else:
        lock_counts[location][lock] = 1

# 結果を表示する
for location, locks in lock_counts.items():
    print("【" + location + "】")
    for lock, count in locks.items():
        print(lock, ":", count, "件")
```

## 場所ごとの施錠状況の割合

場所ごとに施錠していた場合と施錠していなかった場合の割合を計算する。
件数だけでは場所ごとの違いが分かりにくいため、それぞれの場所の合計を基準にして割合を比較する。

```python
for location, locks in lock_counts.items():
    location_total = sum(locks.values())

    print("【" + location + "】")

    for lock, count in locks.items():
        percentage = count / location_total * 100
        print(f"{lock}：{count}件（{percentage:.1f}%）")

    print()
```

## 施錠状況の比較グラフ

発生場所ごとの施錠状況を積み上げ棒グラフで表示する。場所ごとに施錠状況の割合を比較しやすくするため、各場所の合計を100%として表示する。

```python
# 発生場所ごとの施錠状況を文字で表示する

for location, locks in lock_counts.items():
    total = sum(locks.values())

    print("【" + location + "】")

    for lock, count in locks.items():
        percentage = count / total * 100
        bar = "■" * int(percentage / 5)

        print(f"{lock:12} | {bar} {percentage:.1f}%")

    print()
    
```

## 分析結果と考察

今回の分析では、発生場所ごとの件数を比較したところ、最も多かったのは、その他の場所を除いて、駐車（輪）場が、3090件だった。

また、発生場所ごとの施錠状況を比較すると、一戸建住宅では施錠せずに盗難に遭った割合が82.4％と最も高く、その他の住宅や駐車場でも施錠せずに盗難に遭った割合が6割以上を占めていた。一方、4階建て以上の共同住宅では施錠した割合が42.7％と、ほかの場所に比べて高い傾向が見られた。

このことから、自転車盗難は発生場所によって件数や施錠状況に違いがあることが分かった。特に一戸建住宅や駐車（輪）場など、施錠せずに盗難に遭った割合が高い場所に注目する必要があると考えた。これらの場所では、鍵をかける習慣を徹底することが重要だと考える。

## 防犯対策の提案

今回の分析結果をもとに、発生場所に合わせた防犯対策が必要だと考えた。

例えば、駐車（輪）場では盗難件数が3,090件と多く、そのうち64.6％が施錠せずに盗難に遭っていた。そのため、利用者に施錠を呼びかける看板を設置したり、二重ロックを推奨したりする対策が考えられる。

また、一戸建住宅では施錠せずに盗難に遭った割合が82.4％と特に高かった。そのため、自宅の敷地内に自転車を置く場合でも、鍵をかける習慣をつけることが重要だと考える。

一方で、施錠していた自転車も盗難に遭っていることから、鍵をかけるだけでなく、防犯カメラの設置や人目につきやすい場所への駐輪など、周囲の環境を整えることも大切だと考えた。
また、施錠していた自転車の盗難も記録されていることから、鍵をかけるだけでなく、駐輪する場所の環境にも注意することが大切だと考える。

自転車を利用する人が施錠を徹底することに加えて、駐輪場の管理や防犯設備など、周囲の環境を整えることも盗難防止につながると考えた。

## 分析の限界  
今回のデータは自転車盗難の発生件数をまとめたものであり、それぞれの場所に何台の自転車が駐輪されていたかは分からない。そのため、件数が多い場所ほど盗難に遭う確率が高いとは言い切れない。また、施錠状況だけで盗難の原因を判断することもできない。

```python
# 発生場所ごとの件数の合計を確認する
location_total = sum(location_counts.values())

print("元のデータ件数：", len(incidents))
print("発生場所ごとの集計件数：", location_total)

if location_total == len(incidents):
    print("集計結果は元のデータ件数と一致しています。")
else:
    print("件数が一致していません。データを確認してください。")
```
