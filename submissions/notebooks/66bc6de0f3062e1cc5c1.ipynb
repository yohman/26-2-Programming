### 大きい順に並び替えて分布を調べる ###

```python
import pandas as pd
df = pd.read_csv('japan.csv')
# 可住地面積 / 総面積（北方地域及び竹島を除く） が大きい順に値を表示する
for index, row in df.iterrows():
    results = []

    for index, row in df.iterrows():
        #計算する
        zinko = row['可住地面積'] / row['総面積（北方地域及び竹島を除く）']
        #結果をリストに追加する
        results.append((zinko, row['都道府県']))
# 可住地面積 / 総面積（北方地域及び竹島を除く） が大きい順に値を表示する
    for zinko, prefecture in sorted(results, reverse=True):
        print(f"{prefecture}: {zinko:.2%}")
    results.append(zinko)
# 値を大きい順に表示する

```

結果を見ると、前回の結果でも沖縄の49.35という50%に限りなく近い値を見切ってしまうあたりやはり基準として失敗だと思った。

### 結果のグラフを作成する ###

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('japan.csv')
for index, row in df.iterrows():
    results = []

    for index, row in df.iterrows():
        #計算する
        zinko = row['可住地面積'] / row['総面積（北方地域及び竹島を除く）']
        results.append((zinko))


plt.figure(figsize=(8, 4))
plt.hist(results, bins=8, color="#315876", edgecolor="white")
plt.xlabel("kajuutimennseki / soumennseki")
plt.ylabel("Number of prefectures")
plt.title("Distribution of kajuutimennsekihi by Prefecture")
plt.tight_layout()
plt.show()


```
