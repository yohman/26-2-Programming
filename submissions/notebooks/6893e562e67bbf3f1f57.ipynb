## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# 辞書（dict）：Botのプロフィールを「名前: 値」の形で入れる
bot = {
    "name": "シャドウクロウ",             # Botの名前
    "role": "夜に突然出現するペストマスクの男",  # botの役割
    "likes": ["胡散臭い骨董品", "深夜に徘徊", "かぼちゃ料理"],  # リスト（list）：好きなもの三つ
}

# 気分ごとの顔。キーは normal / happy / tired の三つ
faces = {
    "normal": "🎭",
    "happy": "🎭✨",
    "tired": "🎭💤",
    "angry": "🎭💢"
}

# プロフィールを表示する
print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random  # ランダムに選ぶ道具を使えるようにする

# 気分ごとの返事リスト（自分らしい言葉に変えよう）
responses = {
    "normal": [
        f"おや…どうかなさいましたか？私の名は{bot['name']}。毎夜徘徊している者ですよ。",
        "そのお話、詳しく聞かせてください。",
        f"私は{bot['likes'][0]}が好きでしてね。あなたはどうでしょう？",
        "ハロウィンの夜は奇妙な体験でいっぱいですから。くれぐれもお気をつけて。",
    ],
    "happy": [
        "いいですね。その喜び、是非私にも分けていただきたいものです。",
        "面白いですね。そのお話、もう少し詳しく聞かせてください。",
        "そこまでうれしかったのですね。実に興味深い…",
        f"私なら{bot['likes'][1]}して、もっとたくさんの話を聴いてみたいものです。",
    ],
    "tired": [
        "だいぶ疲れているようですね。私の影で回復でもするとしますか？",
        "あわてる必要などありません。リラックスも作業の内ですよ。",
        "全てではなく数％を目指すだけでもいいのですよ。",
        f"{bot['likes'][2]}でも食べて回復するとよいですよ。",
    ],
    "angry": [
        "相当怒りの感情がたまっているようですね。この影であればひと払いで…",
        "その怒りの感情、私の影の中に収めなさい。気分が楽になるでしょう。",
        "でしたら、この秘薬を飲めば治まるかもしれませんよ。",
        "そんな時こそ、深く眠ることが大切です。",
    ],
}

# 確認：最初の返事と、ランダムな返事を表示
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 1. 気分を判定する言葉のリスト（自分で増やしてみよう）
# 英語もOK（英語は小文字で書く）
happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "最高",
    "ハッピーハロウィン！！！"
    "happy",
    "good",
    "great",
    "fun",
    "fine",
    "happy Halloween!!!"
]
tired_words = [
    "元気じゃない",
    "つかれ",
    "疲れ",
    "眠い",
    "しんど",
    "つらい",
    "tired",
    "sleepy",
    "sad",
    "bad",
    "not good",
    "not happy",
]
angry_words = [
    "怒", 
    "ムカつく", 
    "むかつく", 
    "イライラ", 
    "おこ", 
    "腹立つ", 
    "angry",
    "bad",
    "upset",
]

# 2. Botが質問する
question = "今宵の気分はいかがですか？ 元気ですか？ それとも…お疲れですか？"
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える（入力が中断された場合は空欄として扱う）
try:
    message = input(f"{faces['normal']} {question}").strip().lower()  # lower() で小文字にそろえる
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

# ループで言葉をチェックして気分を上書きしていく
for word in happy_words:
    if word in message:
        mood = "happy"

for word in tired_words:
    if word in message:
        mood = "tired"

for word in angry_words:
    if word in message:
        mood = "angry"

# 5. 気分に合う返事をランダムに選んで表示する
reply = random.choice(responses[mood])
print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
