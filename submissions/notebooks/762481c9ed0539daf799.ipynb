## Botの紹介
お腹がすいた学生のために、おすすめの学食メニューや今の気分に合わせた返しをしてくれるボットです。

## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# 1 | Botのプロフィールと表情設定

# プロフィール設定（辞書とリスト）
bot = {
    "name": "がくしょくbot",
    "role": "キャンパス案内係",
    "likes": ["ラーメン", "カレー", "唐揚げ定食"]
}

# 気分ごとの顔文字設定
faces = {
    "normal": "(・∀・)",
    "happy": "(^o^)",
    "tired": "(´・ω4・`)"
}

# プロフィールの表示
print(f"名前: {bot['name']} / 役割: {bot['role']}")
print(f"好きなメニュー: {bot['likes'][0]}, {bot['likes'][1]}, {bot['likes'][2]}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
# 2 | 三つの気分と会話システム
import random

# 気分ごとの返答リスト（3パターン以上）
responses = {
    "happy": [
        "今日は絶好調！一番高い『唐揚げスペシャル』を食べちゃおう！",
        "元気が一番！今日も学食のカレーは大盛りで行こう！",
        "最高だね！食後にアイスも買っちゃおう！",
        "いいね！今日も良い一日になりそうだ！"
    ],
    "tired": [
        "お疲れ様！あたたかい『ラーメン』を食べて回復しよう。",
        "無理しちゃダメだよ。甘い飲み物でも買って休憩してね。",
        "ふぁ〜あ...僕も眠いよ。学食で一息つこう。",
        "今日もお疲れ様。ゆっくり休んでね。"
    ],
    "normal": [
        "いつもの日替わり定食がおすすめだよ！",
        "学食は12時過ぎると混むから早めに行こうね。",
        "うどんかそばでサクッと済ませるのもアリだね。",
        "今日も一日マイペースに頑張ろう！"
    ]
}

print(f"{bot['name']} {faces['normal']}: やあ！話しかけてね。（『終了』や『バイバイ』と入力すると終わります）\n")

# 会話のメインループ
while True:
    user_input = input("あなた: ")
    
    # 1. 会話を終了する方法
    if user_input == "終了" or user_input == "バイバイ" or user_input == "exit":
        print(f"{bot['name']} {faces['normal']}: またね！学食で待ってるよー！")
        break
        
    # 2. 条件分岐（if / elif / else）による返答切り替え
    elif "元気" in user_input or "嬉" in user_input or "最高" in user_input:
        reply = random.choice(responses["happy"])
        print(f"{bot['name']} {faces['happy']}: {reply}\n")
        
    elif "疲" in user_input or "眠" in user_input or "つらい" in user_input:
        reply = random.choice(responses["tired"])
        print(f"{bot['name']} {faces['tired']}: {reply}\n")
        
    elif "普通" in user_input or "お腹" in user_input or "おすすめ" in user_input:
        reply = random.choice(responses["normal"])
        print(f"{bot['name']} {faces['normal']}: {reply}\n")
        
    # 3. 想定外の入力への返事
    else:
        print(f"{bot['name']} {faces['normal']}: うーん、ちょっと何言ってるかわからないや！『疲れた』とか『おすすめ』って言ってみて！\n")
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 1. 気分を判定する言葉のリスト（自分で増やしてみよう）
# 英語もOK（英語は小文字で書く）
happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "最高",
    "happy",
    "good",
    "great",
    "fun",
    "fine",
]
tired_words = [
    "元気じゃない",
    "つかれた",
    "疲れ",
    "眠い",
    "しんど",
    "つらい",
    "tired",
    "sleepy",
    "sad",
    "bad",
    "not good",
    "not happy",
]

# 2. Botが質問する
question = "今日の気分はどう？ 元気？ つかれた？ (How are you?)"
print(f"{faces['normal']} {bot['name']} : {question}")

# 3. ユーザーが答える（入力が中断された場合は空欄として扱う）
try:
    message = input(f"{faces['normal']} {question}").strip().lower()  # lower() で小文字にそろえる
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた : {message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

# ループは準備済み：happy の言葉があれば happy にする
for word in happy_words:
    if word in message:
        mood = "happy"
        break

# tired の言葉があれば tired にする（「元気じゃない」などを判定）
for word in tired_words:
    if word in message:
        mood = "tired"
        break

# 5. 条件分岐（if / elif / else）で返事を決めて出力する
if mood == "happy":
    reply = random.choice(responses["happy"])
    print(f"{faces['happy']} {bot['name']} : {reply}")
elif mood == "tired":
    reply = random.choice(responses["tired"])
    print(f"{faces['tired']} {bot['name']} : {reply}")
else:
    reply = random.choice(responses["normal"])
    print(f"{faces['normal']} {bot['name']} : {reply}")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
