# My Chatbot｜今日食べたいものを相談しよう
このBotは、気分に合わせて食べものをおすすめし、食事選びを楽しくするためのBotです。

## 1｜Botのプロフィール
食べものがテーマのキャラクターにします。辞書に名前・役割を、リストに好きな食べものを三つ入れます。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。


```python
# 辞書（dict）：Botのプロフィール
bot = {
    "name": "もぐもぐ",             # Botの名前
    "role": "食べもの選びの相談相手",  # Botの役割
    "likes": ["ラーメン", "お寿司", "カレー"],  # 好きな食べもの三つ
}

# 気分ごとの顔
faces = {
    "normal": "🍽️",
    "happy": "😋",
    "tired": "🥣",
}

# プロフィールを表示する
print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つに分け、食べものに関する返事を用意します。

- `responses["happy"][0]`：元気なときの最初の返事。
- `random.choice(responses["happy"])`：元気なときの返事から一つランダムに選びます。


```python
import random  # ランダムに選ぶ道具を使えるようにする

# 気分ごとの返事リスト
responses = {
    "normal": [
        f"こんにちは！{bot['name']}だよ。今日は何を食べたい？",
        "あっさりしたものと、こってりしたものならどっちがいい？",
        f"私は{bot['likes'][0]}が好き！あなたの好きな食べものは？",
        "今日のごはんを一緒に考えよう！",
    ],
    "happy": [
        "元気いっぱいだね！好きなものを食べて楽しもう！",
        "いいね！今日はちょっと豪華なごはんにする？",
        f"気分がいい日は{bot['likes'][1]}もおすすめだよ！",
        "デザートも食べたくなっちゃうね！",
    ],
    "tired": [
        "おつかれさま。温かいスープやうどんはどう？",
        "疲れたときは、無理せずゆっくりごはんを食べよう。",
        f"食べやすいものなら{bot['likes'][2]}はどうかな？",
        "まずは水分をとって、ひと休みしよう！",
    ],
}

# 確認：最初の返事と、ランダムな返事を表示
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分チェッカー
食べものについて相談するため、入力された言葉から気分を判定して返事をします。一回の実行で一回返事をします。

「元気」「おいしい」「疲れた」「おなかすいた」などを試してみましょう。


```python
# 1. 気分を判定する言葉のリスト
happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "おいしい",
    "美味しい",
    "最高",
    "happy",
    "good",
]
tired_words = [
    "元気じゃない",
    "疲れ",
    "つかれ",
    "眠い",
    "しんど",
    "おなかすいた",
    "腹減った",
    "お腹すいた",
    "tired",
    "sleepy",
    "sad",
]

# 2. Botが質問する
question = "今の気分は？ 元気？ 疲れた？ おなかすいた？"
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える
try:
    message = input(f"{faces['normal']} {question} ").strip().lower()
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

# 元気な言葉があれば happy にする
for word in happy_words:
    if word in message:
        mood = "happy"

# 疲れた・おなかがすいた等の言葉は tired にする
# 「元気じゃない」には「元気」も含まれるので、こちらを後に調べます
for word in tired_words:
    if word in message:
        mood = "tired"

# 5. 気分に合う返事をランダムに選んで表示
reply = random.choice(responses[mood])
print(f"{faces[mood]} {bot['name']}：{reply}")
```

 > **ポイント：** 「元気じゃない」には「元気」も含まれます。先に happy と判定されても、後のループで tired に上書きされるようにしています。言葉を追加するときは、意図した気分になるか確認しましょう。


## 4｜自分のキャラクターにする
