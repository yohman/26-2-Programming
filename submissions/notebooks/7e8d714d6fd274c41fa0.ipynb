## 調べたいこと

阿見町の朝6時と昼14時では、予報の気温は何℃違う？

出典： Open-Meteo (https://open-meteo.com/)10月21日アクセス


````python
import json
from urllib.request import urlopen

url = "https://api.open-meteo.com/v1/forecast?latitude=36.0215&longitude=140.2244&hourly=temperature_2m&timezone=Asia%2FTokyo"

with urlopen(url, timeout=20) as response:
    data = json.load(response)
````

## 気温を取り出す

data は辞書。その中の hourly に、時刻と気温のリストがあります。
0時が添字0なので、6時は [6]、14時は [14] です。

````python
temperatures = data["hourly"]["temperature_2m"]
morning = temperatures[6]
afternoon = temperatures[14]

print(f"対象日: {data['hourly']['time'][0][:10]}")
print(f"朝6時: {morning} ℃")
print(f"昼14時: {afternoon} ℃")
print(f"昼と朝の差: {afternoon - morning:.1f} ℃")
````

## 分かったこと

10月21日の予報では、朝6時は18.9℃、昼14時は27.8℃。昼の方が8.9℃高かった。

確認： 元のJSONの temperature_2m で、添字6が18.9、添字14が27.8であることを確かめた。

限界： これは予報であり、実測値ではない。風や雨は調べていない。取得日が変わると値も変わるので、再実行したら観察も書き直す。
