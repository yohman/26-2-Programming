```python
import random

print("=== 今日の気分でコーデBot ===")
print("今日の気分を入力してね！")
print("終了したいときは「終了」と入力してね。")

while True:
    mood = input("\n今日の気分は？（おしゃれ・カジュアル・元気・リラックス）：")

    if mood == "終了":
        print("今日もおしゃれを楽しんでね！またね！")
        break

    elif mood in ["おしゃれ", "大人っぽい"]:
        outfits = [
            "テーラードジャケットにスラックスを合わせて、大人っぽく！",
            "シャツとワイドパンツで、きれいめコーデに！",
            "シンプルなニットとスラックスで、落ち着いた印象に！"
        ]
        print(random.choice(outfits))

    elif mood in ["カジュアル", "楽な服"]:
        outfits = [
            "パーカーとデニムで、リラックスしたコーデに！",
            "無地のTシャツとワイドパンツで、シンプルに！",
            "スウェットとスニーカーで、動きやすい服装に！"
        ]
        print(random.choice(outfits))

    elif mood in ["元気", "明るい"]:
        outfits = [
            "明るい色のトップスとデニムで、爽やかに！",
            "ボーダーTシャツとチノパンで、軽やかな印象に！",
            "有彩色を入れて、気分が上がるコーデに！"
        ]
        print(random.choice(outfits))

    elif mood in ["疲れた", "リラックス"]:
        outfits = [
            "柔らかい素材のトップスとゆったりしたパンツで、楽な服装に！",
            "カーディガンとストレートパンツで、落ち着いたコーデに！",
            "スウェットとスニーカーで、リラックスできる服装に！"
        ]
        print(random.choice(outfits))

    else:
        print("その気分はまだ勉強中！「おしゃれ」「カジュアル」「元気」「疲れた」から選んでみてね！")
```

```python
import random

print("================================")
print("      今日の気分でコーデBot")
print("================================")
print("気分と気温から、おすすめの服装を提案するよ！")
print("終了したいときは「終了」と入力してね。")

while True:
    mood = input("\n今日の気分は？（おしゃれ・カジュアル・元気・リラックス）：")

    if mood == "終了":
        print("今日もおしゃれを楽しんでね！またね！")
        break

    elif mood not in ["おしゃれ", "カジュアル", "元気", "リラックス"]:
        print("その気分はまだ勉強中！別の気分を入力してね。")
        continue

    try:
        temperature = float(input("今日の気温は何℃？："))
    except ValueError:
        print("気温は数字で入力してね！")
        continue

    if temperature < -50 or temperature > 50:
        print("気温は-50℃から50℃の範囲で入力してね！")
        continue

    # 気温に合わせて服装を選ぶ
    if temperature < 5:
        clothes = [
            "厚手のコートとセーター、マフラーで暖かく！",
            "ダウンジャケットと長ズボンで防寒しよう！",
            "冬用コートに手袋を合わせて寒さ対策！"
        ]
    elif temperature < 11:
        clothes = [
            "コートとニットを合わせた暖かいコーデ！",
            "ジャケットにセーターを重ね着しよう！",
            "厚手のアウターと長ズボンがおすすめ！"
        ]
    elif temperature < 16:
        clothes = [
            "テーラードジャケットと長袖シャツ！",
            "ニットとスラックスで落ち着いたコーデ！",
            "軽めのコートと長袖トップスを合わせよう！"
        ]
    elif temperature < 21:
        clothes = [
            "カーディガンと長袖シャツで快適に！",
            "薄手のジャケットとデニムを合わせよう！",
            "長袖シャツとチノパンですっきりと！"
        ]
    elif temperature < 26:
        clothes = [
            "長袖シャツと軽めのパンツで爽やかに！",
            "薄手のトップスとデニムでシンプルに！",
            "半袖の上に薄手の羽織りを合わせよう！"
        ]
    else:
        clothes = [
            "半袖Tシャツと通気性のよいパンツ！",
            "涼しい素材のシャツと軽いパンツ！",
            "風通しのよい服装で暑さ対策をしよう！"
        ]

    # 気分に合わせてスタイルを選ぶ
    if mood == "おしゃれ":
        styles = [
            "色を落ち着いたトーンでまとめて、大人っぽく！",
            "アクセサリーやバッグを取り入れて、おしゃれに！",
            "色の組み合わせを工夫して、きれいめに！"
        ]
    elif mood == "カジュアル":
        styles = [
            "デニムやスニーカーを合わせてカジュアルに！",
            "シンプルな色合いで気軽に楽しもう！",
            "動きやすいアイテムでラフな雰囲気に！"
        ]
    elif mood == "元気":
        styles = [
            "明るい色を取り入れて、爽やかに！",
            "お気に入りの色を主役にしてみよう！",
            "スニーカーを合わせて軽快な印象に！"
        ]
    else:
        styles = [
            "ゆったりした服を選んでリラックス！",
            "柔らかい素材を使って、楽なコーデに！",
            "落ち着いた色合いで、ゆるい雰囲気に！"
        ]

    print("\n---------- おすすめコーデ ----------")
    print(f"今日の気分：{mood}")
    print(f"今日の気温：{temperature}℃")
    print("気温に合わせた服装：" + random.choice(clothes))
    print("気分に合わせたスタイル：" + random.choice(styles))
    print("------------------------------------")
```
