## 1．データの読み込み

授業で配布された japan.csv を、Python標準の csv を使って読み込む。

```python
import csv

with open("japan.csv", encoding="utf-8-sig", newline="") as file:
    data = list(csv.DictReader(file))

print(len(data))
```

## 2．データの確認

「都道府県」「総人口」「65歳以上人口」の3列を使う。最初の3件を表示してデータの中身を確認する。

```python
print("列数", len(data[0]))
print("都道府県 ｜ 総人口（人） ｜ 65歳以上人口（人）")
for row in data[:3]:
    print(row["都道府県"], "｜", row["総人口"], "｜", row["65歳以上人口"])
```

## 3．割合の計算と比較

都道府県ごとに「65歳以上人口 ÷ 総人口 × 100」で割合を求める。最初の3件の割合と、30％以上だった都道府県の数を表示する。

```python
results = []
over_30 = 0

for row in data:
    population = int(row["総人口"])
    seniors = int(row["65歳以上人口"])
    share = seniors / population * 100

    results.append({
        "都道府県": row["都道府県"],
        "人数": seniors,
        "割合": share,
    })

    if share >= 30:
        over_30 += 1

print("都道府県 ｜ 65歳以上の割合（最初の3件）")
for row in results[:3]:
    print(f'{row["都道府県"]} ｜ {row["割合"]:.2f}％')

print(f"65歳以上の割合が30％以上　{over_30}/{len(results)}都道府県")
```

## 4．結果

65歳以上の割合が高い順と、人数が多い順に、それぞれ上位5件を表示して比べる。

```python


by_share = sorted(results, key=lambda row: row["割合"], reverse=True)[:5]
by_count = sorted(results, key=lambda row: row["人数"], reverse=True)[:5]

for title, ranking in [("割合の上位5件", by_share), ("人数の上位5件", by_count)]:
    print()
    print(title)
    print("順位 ｜ 都道府県 ｜ 65歳以上の割合 ｜ 65歳以上の人数")
    for rank, row in enumerate(ranking, start=1):
        print(f'{rank} ｜ {row["都道府県"]} ｜ {row["割合"]:.2f}％ ｜ {row["人数"]:,}人')


```

## 5．わかったことと上司への提案

65歳以上の割合は秋田県が38.60％で最も高く、人数は東京都が320万2千人で最も多かった。割合と人数のどちらを見るかで、最初に調べる地域が変わることがわかった。

秋田県では窓口までの移動時間を調べて出張相談を検討し、東京都では相談件数や待ち時間を調べて予約枠の見直しを検討したい。このデータは都道府県単位なので、実際の計画には市町村別の人口や窓口の利用記録も必要になる。
