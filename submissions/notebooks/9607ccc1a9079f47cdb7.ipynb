## 発見したこと・問い
* **問い**: 今週の東京の時間別気温データにおいて、最高気温と最低気温の差はどのくらいあるか？
* **発見したこと**: 直近1週間のデータから最高気温と最低気温を比較したところ、時間帯によって大きな寒暖差が存在することが分かった。

## 元データと照らして確かめた値
* APIの `data["hourly"]["temperature_2m"]` リストから`max()` と `min()` 関数を使って最高気温・最低気温の数値を算出して比較・確認しました。

## データの限界
* 気温データの変動要因となる「降水量」「風速」「日照時間」などの他の気象要素が含まれていないため気温が変化した詳しい理由までは特定できない点。

## 気温を取り出す

`data`は辞書。その中の`hourly`に、時刻と気温のリストがあります。

0時が添字0なので、8時は`[8]`、14時は`[14]`です。

```python
import json
from urllib.request import urlopen

# 東京の1週間分の時間別気温データを取得
url = "https://api.open-meteo.com/v1/forecast?latitude=35.6895&longitude=139.6917&hourly=temperature_2m&timezone=Asia%2FTokyo"

with urlopen(url, timeout=20) as response:
    data = json.load(response)

# 気温データのリストを取り出し
temperatures = data["hourly"]["temperature_2m"]

# 最高・最低・平均・差の計算
max_temp = max(temperatures)
min_temp = min(temperatures)
avg_temp = sum(temperatures) / len(temperatures)
diff_temp = max_temp - min_temp

# 要約結果の出力
print("=== 東京の週間天気データ分析結果 ===")
print(f"取得データ件数 : {len(temperatures)}時間分")
print(f"最高気温       : {max_temp} ℃")
print(f"最低気温       : {min_temp} ℃")
print(f"平均気温       : {avg_temp:.1f} ℃")
print(f"最大寒暖差     : {diff_temp:.1f} ℃")
```

東京の週間天気データを分析したところ以下の数値が分かった。

* **最高気温**: 28.7 ℃
* **最低気温**: 12.7 ℃
* **平均気温**: 18.6 ℃
* **最大寒暖差**: 16.0 ℃

直近1週間の中で気温が最も低い時間帯と最も高い時間帯では 16.0 ℃ もの大きな気温差が存在することが確認できた。

**確認**: 元のJSONの `temperature_2m` リストから`max()` と `min()` を用いて数値を取り出し、差を算出して確かめた。

**限界**: これは予報データであり実測値ではない点、また風速や湿度など他の気象要素を含んでいない点。
