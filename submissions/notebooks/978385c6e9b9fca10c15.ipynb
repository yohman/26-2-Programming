## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# ヤドンのプロフィール
bot = {
    "name": "ヤドン",
    "role": "のんびりしたポケモン",
    "likes": ["お昼寝", "しっぽを水につけること", "ぼーっとすること"],
}

# 気分ごとの顔
faces = {
    "normal": "(・～・)",
    "happy": "(＾～＾)",
    "tired": "(－～－)",
}

print(f"{faces['normal']} {bot['name']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random

# 気分ごとの返事
responses = {
    "normal": [
        "やぁ～……ぼく、ヤドンだよぉ～。",
        "ん～……何のお話だっけぇ～？",
        "ぼ～っとしてたら、もうこんな時間だぁ～。",
        f"ぼくは{bot['likes'][0]}が好きなんだぁ～。",
        "ん～……ちょっと考えるねぇ～。",
    ],
    "happy": [
        "やったぁ～……なんだかうれしいねぇ～。",
        "えへへぇ～……ぼくもうれしいよぉ～。",
        "いいことがあったんだねぇ～……よかったぁ～。",
        "ん～……お祝いにお昼寝しよっかぁ～。",
        "わぁ～……すごいねぇ～！",
    ],
    "tired": [
        "ん～……疲れちゃったのぉ～？",
        "ぼくと一緒にお昼寝しよぉ～……。",
        "無理しなくてもいいんだよぉ～。",
        "ふわぁ～……ぼくも眠くなってきたぁ～。",
        "ゆっくり休んでねぇ～……。",
    ],
}

# 返事を確認
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 元気な言葉
happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "最高",
    "好き",
    "かわいい",
    "勝った",
    "成功",
    "happy",
    "good",
]

# 疲れた言葉
tired_words = [
    "元気じゃない",
    "疲れた",
    "疲れ",
    "つかれ",
    "眠い",
    "しんどい",
    "つらい",
    "悲しい",
    "負けた",
    "失敗",
    "嫌い",
    "tired",
    "sad",
    "not good",
]

# ヤドンが質問する
question = "やぁ～……今日はどんな気分なのぉ～？"

print(f"{faces['normal']} {bot['name']}：{question}")

# ユーザーが答える
try:
    message = input(f"{faces['normal']} {question}").strip().lower()
except KeyboardInterrupt:
    message = ""

print(f"あなた：{message}")

# 最初は普通の気分
mood = "normal"

# 元気な言葉を探す
for word in happy_words:
    if word in message:
        mood = "happy"

# 疲れた言葉を探す
for word in tired_words:
    if word in message:
        mood = "tired"

# ヤドンの返事
reply = random.choice(responses[mood])

print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
