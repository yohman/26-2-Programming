# 発展　自転車レンタル経営ゲーム

ナンバーごとに季節や天気を命名すること。
import random で日付をランダムに選ぶ機能を付けること。
-準備個数10ポイントした後、if準備個数>cut ならcut*20pointすればよくて,if 準備個数<cut なら、準備個数20 すること
までは自分で書くことができた。

AIにケアレスミスを指摘、セーブ機能を追加してもらった。while money >= 10:　　　「while」はループの中でゲームを繰り返しされ次のゲームでも使い続けられるとのこと。money += incomeでこのゲームないでは＋され続ける、ゲームを続けるの機能をそもそも忘れているなど。

```python
import random

# 季節・天候の日本語表記
seasons = {
    1: "春",
    2: "夏",
    3: "秋",
    4: "冬"
}

weather = {
    1: "晴れ・薄曇り",
    2: "霧・曇り",
    3: "小雨・小雪など",
    4: "大雨・大雪など"
}

# 初期所持金
money = 50000

print("自転車レンタル経営ゲーム")
print("初期所持金:", money, "ポイント")

while money >= 10:

    # CSVからランダムに1日選ぶ
    random_index = random.randint(0, len(df) - 1)
    day = df.iloc[random_index]

    # その日の状況を表示
    print("\n--------------------")
    print("日付:", day["dteday"])
    print("季節:", seasons[int(day["season"])])
    print("天候:", weather[int(day["weathersit"])])
    print("風の強さ（データ上の指数）:", round(day["windspeed"], 2))

    print("現在の所持金:", money, "ポイント")

    # 用意する自転車の台数
    try:
        prepared = int(input("用意する自転車の台数: "))
    except ValueError:
        print("整数を入力してください。")
        continue

    if prepared < 0:
        print("0以上の台数を入力してください。")
        continue

    # 準備費用
    cost = prepared * 15

    if cost > money:
        print("所持金が足りません。")
        continue

    money -= cost

    # 実際の需要
    cut = int(day["cnt"])

    # 貸し出せた台数
    if prepared > cut:
        rented = cut
    else:
        rented = prepared

    # 売上
    income = rented * 17.5
    money += income

    # 結果発表
    print("\n実際の需要:", cut, "台")
    print("貸し出せた台数:", rented, "台")
    print("準備費用:", cost, "ポイント")
    print("売上:", income, "ポイント")
    print("現在の所持金:", money, "ポイント")

    # 続けるかどうか
    answer = input("次の日も挑戦しますか？（y/n）: ")

    if answer.lower() != "y" and answer.lower() != "yes":
      break

print("\nゲーム終了")
print("最終所持金:", money, "ポイント")
```
