## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
import random

# 1. 辞書（dict）：Botのプロフィール
bot = {
    "name": "ぽち",             # Botの名前
    "role": "お世話できるペット",  # Botの役割
    "likes": ["肉", "ワンチュール", "骨"],  # リスト（list）：好きなもの三つ
}

# 2. 気分ごとの顔（キーは normal / happy / angry / sleep）
faces = {
    "normal": "🐶",
    "happy": "🐶✨",  # にこにこ
    "angry": "🐶💢",  # おこ・かみつく
    "sleep": "🐶💤",  # ねころぶ
    "tired": "🐶💢",
}

# 最初におこった顔を表示（※おこった顔を見せたい場合は 'angry' に変更）
print(f"{faces['angry']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
responses = {
    "normal": [ # なでられた時（ねころぶ・なでなで）
        "ん〜、気持ちいいワン…（ゴロンと横になった）",
        "気持ちよくてウトウトしてきたワン…",
        f"なでなで大好き！お返しのなでなでワン！",
    ],
    "happy": [  # えさをあげた時（にこにこ大喜び）
        f"わーい！大好物の {bot['likes'][0]} だ！おいしいワン！",
        "もっとちょうだい！",
        f"たりねー {bot['likes'][1]} よこせ",
    ],
    "tired": [  # たたかれた時（かみつく・ガルル）
        "ガルルル…ッ！！ガブッ！！（痛くて噛みつかれた！）",
        "クーンこれだから人間は",
        "キャンキャン！痛いからやめてワン！",
    ],
}
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 1. 行動を判定する言葉のリスト
happy_words = ["えさ", "餌", "ごはん", "エサ", "food", "feed"]  # えさをあげる言葉
tired_words = ["たたく", "叩く", "パンチ", "ぶつ", "hit", "punch"]  # たたく言葉
normal_words = ["なでる", "撫でる", "なでなで", "撫で撫で", "pet", "pat"]  # なでる言葉

# 2. Botが話しかける
question = "「えさ」「たたく」「なでる」のどれかをしてね！（「ハウス」で終了）"
print(f"{faces['normal']} {bot['name']}：{question}")

# 繰り返し会話する仕組み（whileループ）
while True:
    # 3. ユーザーが入力する
    try:
        message = input(f"{faces['normal']} あなた：").strip().lower()
    except KeyboardInterrupt:
        message = ""

    # 会話を終了する条件（「ハウス」や「はうす」で終了）
    if message == "ハウス" or message == "はうす" or message == "house":
        print(f"{faces['sleep']} {bot['name']}：クーン…（トコトコ家に入っていった。またね！）")
        break

    # 4. 行動を決める（まず normal にしておく）
    mood = "normal"

    # なでる言葉チェック
    for word in normal_words:
        if word in message:
            mood = "normal"

    # えさをあげる言葉チェック
    for word in happy_words:
        if word in message:
            mood = "happy"

    # たたく言葉チェック
    for word in tired_words:
        if word in message:
            mood = "tired"

    # どれにも当てはまらない（想定外の入力）チェック
    is_known_word = False
    for word in happy_words + tired_words + normal_words:
        if word in message:
            is_known_word = True
            break

    # 5. 気分に合う返事を表示する
    if is_known_word:
        reply = random.choice(responses[mood])
        print(f"{faces[mood]} {bot['name']}：{reply}")
    else:
        print(f"{faces['angry']} {bot['name']}：首をかしげている…「えさ」「たたく」「なでる」で話しかけてみてね！")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
