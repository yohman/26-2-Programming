### ■ 何を改良し、なぜそうしたか
1. 地図にタイトルを追加
   - 一目で何の地図であるか（世界の地震分布地図）を把握できるようにするため。
2. 地震マーカーをピンから「円（CircleMarker）」に変更し、マグニチュードに応じた色とサイズに可視化
   - マグニチュードが大きい地震ほど「大きな円・赤い色」で目立つように変更し、危険度を直感的に判別できるようにしました。

### ■ 地図から気づいたこと
- 太平洋火山帯（Ring of Fire）沿いに地震が集中している。
- マグニチュードの大きい地震（赤い大きな円）は特定のエリアに偏っている。

### ■ 変更後の地図が意図どおり動くことを、どう確かめたか
- 各円をクリックした際に、正しいマグニチュードと場所のポップアップが表示されるか確認した。
- 3段階（緑・オレンジ・赤）の色とサイズの変化が正しく表示されているか確認した。

### ■ CSVの出典とダウンロード日
2026年10月7日時点のUSGSデータ

```python
import folium
import pandas as pd

# CSVファイルの読み込み
df = pd.read_csv('all_hour.csv')

# 地図の初期化
m = folium.Map(location=[20, 0], zoom_start=2, tiles='OpenStreetMap')

# タイトルを追加
title_html = '<h3 align="center" style="font-size:20px"><b>【世界の地震分布地図】直近の地震発生状況</b></h3>'
m.get_root().html.add_child(folium.Element(title_html))

# 円マーカーの描画
for idx, row in df.iterrows():
    mag = row['mag']
    lat = row['latitude']
    lon = row['longitude']
    place = row['place']

    if mag >= 5.0:
        color = 'red'
        radius = mag * 3
    elif mag >= 3.0:
        color = 'orange'
        radius = mag * 2
    else:
        color = 'green'
        radius = mag * 1.5

    folium.CircleMarker(
        location=[lat, lon],
        radius=radius,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.6,
        popup=f"場所: {place}<br>マグニチュード: {mag}",
    ).add_to(m)

# 地図の表示
m
```
