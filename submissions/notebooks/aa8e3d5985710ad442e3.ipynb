## 2｜CSVを読む

`latitude`は緯度、`longitude`は経度、`mag`はマグニチュード。CSVの値は文字なので、地図に使う緯度・経度は`float()`で数に変えます。


```python
import csv
import folium

csv_file = "eq.csv"

with open(csv_file, encoding="utf-8-sig") as file:
    earthquakes = list(csv.DictReader(file))

print(f"{len(earthquakes)}件のデータを読みました")
```

## 3｜一行につき、マーカーを一つ

実行する前に、どこにどんな点が出るか予想しよう。`for`の中が各地震に対して繰り返されます。最後の`earthquake_map`でNotebookに地図を表示します。


```python
earthquake_map = folium.Map(location=[20, 0], zoom_start=2)
 
for quake in earthquakes:
    folium.Marker(
        location=[float(quake["latitude"]), float(quake["longitude"])],
        tooltip=f"M {quake['mag']}"
    ).add_to(earthquake_map)
 
earthquake_map
```

```python
import csv
import folium

csv_file = "eq.csv"

with open(csv_file, encoding="utf-8-sig") as file:
    earthquakes = list(csv.DictReader(file))

print(f"{len(earthquakes)}件のデータを読みました")
```

```python
earthquake_map = folium.Map(location=[20, 0], zoom_start=2)

for quake in earthquakes:
    if not quake["mag"]:
        continue

    mag = float(quake["mag"])

    if mag >= 6:
        color = "red"
    elif mag >= 4:
        color = "orange"
    else:
        color = "blue"

    folium.CircleMarker(
        location=[float(quake["latitude"]), float(quake["longitude"])],
        tooltip=f"M {mag}",
        radius=mag * 5,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7
    ).add_to(earthquake_map)

earthquake_map
```

AIで調べたこと
if	最初の条件を確認する
elif	それまでの条件に当てはまらなかった場合、別の条件を確認する
else	どの条件にも当てはまらなかった場合に実行する

改良後のコード（radius=mag * 1.5）magから掛け算で求めることでそれぞれの円の大きさを変えることができるようになった
