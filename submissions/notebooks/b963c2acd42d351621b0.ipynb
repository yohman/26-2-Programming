## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。

```python
# 辞書（dict）：Botのプロフィールを「名前: 値」の形で入れる
bot = {
    "name": "クソガキ",                     # Botの名前
    "role": "なぜか一言多い自称・聞き上手。人をイラつかせるのが得意",  # Botの役割
    "likes": ["正論", "マウント", "煽り"],  # リスト（list）：好きなもの三つ
}

# 気分ごとの顔。キーは normal / happy / tired の三つ
faces = {
    "normal": "😏",
    "happy": "😏✨",
    "tired": "🥱",
}

# プロフィールを表示する
print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal・happy・tired**の三つ。返事の中の `{msg}` は、あとで**あなたが入力した言葉**に置きかえます（これがムカつきポイント）。

```python
import random  # ランダムに選ぶ道具を使えるようにする

# 気分ごとの返事リスト（{msg} = ユーザーが入力した言葉）
responses = {
    "normal": [
        "{msg}、まあまあ落ち着けってｗ",
        "{msg}。うん、うん。で、結論は？",
        "{msg}。おいおいそれしか言えないのかｗ？",
        "{msg}、情けないでちゅねーｗｗｗｗ",
    ],
    "happy": [
        "あ、元気なんだ。よかったね。私は別にどうでもいいけど。",
        "「{msg}」って、そんなにうれしいこと？ まあ、人それぞれだよね。",
        "調子に乗ると痛い目見るよ。あ、応援してるから。",
        f"{bot['likes'][1]}してもいい？ 私のほうが楽しい日、多いと思うけど。",
        "へー、いいなぁ。（でも私のほうが上だけどね）",
    ],
    "tired": [
        "つかれたんだ。大変そうだね〜。私は全然だけど。",
        "「{msg}」か……それ、寝れば治るやつでは？",
        "休めばいいじゃん。……休めない？ ふーん、計画性かな。",
        f"{bot['likes'][2]}されたときの気持ち、ちょっとわかった？",
        "おつかれ。でも、みんなつかれてるからね。",
    ],
    "angry": [
        "なんでそんなに怒ってるの？ 私は全然怒ってないけど。",
        "「{msg}」って、そんなに腹立つこと？ まあ、人それぞれだよね。",
        "怒ると損するよ。あ、応援してるから。",
        f"{bot['likes'][1]}してもいい？ 私のほうが楽しい日、多いと思うけど。",
        "へー、いいなぁ。（でも私のほうが上だけどね）",
    ],
    "sad": [
        "なんでそんなに悲しんでるの？ 私は全然悲しくないけど。",
        "「{msg}」って、そんなに寂しいこと？ まあ、人それぞれだよね。",
        "泣くと損するよ。あ、応援してるから。",
        f"{bot['likes'][1]}してもいい？ 私のほうが楽しい日、多いと思うけど。",
        "へー、いいなぁ。（でも私のほうが上だけどね）",
    ],    
}

# 確認：最初の返事と、ランダムな返事を表示
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」「（何も入力しない）」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

```python
# 1. 気分を判定する言葉のリスト（自分で増やしてみよう）
# 英語もOK（英語は小文字で書く）
happy_words = [
    "元気", "楽しい", "うれしい", "嬉しい", "最高",
    "happy", "good", "great", "fun", "fine", "最高だぜ"
]
tired_words = [
    "元気じゃない", "つかれ", "疲れ", "眠い", "しんど", "つらい",
    "tired", "sleepy", "sad", "bad", "not good", "not happy",
]
angry_words = [
    "むかつき", "腹立", "イラつ", "怒", "むかつく", "腹立つ", "イラつく", "怒る",
    "angry", "mad", "annoyed", "frustrated", "irritated",
]
sad_words = [
    "悲し", "寂し", "泣き", "落ち込", "つらい", "しんどい", "切な", "虚し",
    "sad", "lonely", "crying", "depressed", "painful", "tiring", "heartbreaking", "empty"
]

# 2. Botが質問する
question = "よお　調子どうだい？　俺に会えてうれしいって？　照れるなよ"
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える（入力が中断された場合は空欄として扱う）
try:
    message = input(f"{faces['normal']} {question}").strip().lower()
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

for word in happy_words:
    if word in message:
        mood = "happy"

# 「元気じゃない」は「元気」も含むので、tired を後に調べて上書きする
for word in tired_words:
    if word in message:
        mood = "tired"

# 5. オリジナル機能：何も入力しなかったときは、専用のムカつく返事
if message == "":
    print(f"{faces['normal']} {bot['name']}：……黙ってちゃなんもわかんないよ　ほら、なんか言ってみなよ。")
else:
    # 気分に合う返事をランダムに選び、{msg} をあなたの言葉に置きかえる
    reply = random.choice(responses[mood]).replace("{msg}", message)
    print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
