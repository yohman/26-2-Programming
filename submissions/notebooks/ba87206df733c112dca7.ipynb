# 柏市 週間傘予報（1週間の降水確率チェック）
- **チームメンバー**: しんご、ゆずき、ひなた
- **使う人**: 柏市に住んでいる・通学している学生や社会人
- **問い**: 「今週の柏市は、どの日に傘を持っていくべき？」
- **出典**: [Open-Meteo Weather Forecast API](https://open-meteo.com/en/docs)（タイムゾーン: Asia/Tokyo）

```python
import urllib.request
import json

# 1. 都市名（Kashiwa）から緯度・経度を取得する
city = "Kashiwa"
geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"
geo_res = urllib.request.urlopen(geo_url)
geo_data = json.loads(geo_res.read().decode("utf-8"))

lat = geo_data["results"][0]["latitude"]
lon = geo_data["results"][0]["longitude"]
name = geo_data["results"][0]["name"]
country = geo_data["results"][0]["country"]

# 2. 柏市の7日間の最大降水確率（%）を取得する（日本時間指定）
weather_url = (
    f"https://api.open-meteo.com/v1/forecast?"
    f"latitude={lat}&longitude={lon}"
    f"&daily=precipitation_probability_max"
    f"&timezone=Asia/Tokyo"
)
weather_res = urllib.request.urlopen(weather_url)
weather_data = json.loads(weather_res.read().decode("utf-8"))

dates = weather_data["daily"]["time"]
probs = weather_data["daily"]["precipitation_probability_max"]

# 3. forループで1日ずつ見やすく表示する
print(f"📍 {name} ({country}) の1週間の降水確率予報")
print("-" * 45)

for i in range(len(dates)):
    date = dates[i]
    prob = probs[i]
    
    # None（データなし）対策
    if prob is None:
        print(f"{date} : データなし")
        continue

    # 降水確率に応じて文字の棒（■）を作る（10%ごとに1個）
    bar = "■" * (prob // 10)
    
    # 確率に応じたアドバイス
    if prob >= 50:
        advice = "☔ 傘が必要になりそう"
    elif prob >= 30:
        advice = "🌂 折りたたみ傘があると安心"
    else:
        advice = "☀️ 傘はいらなそう"
        
    print(f"{date} | {prob:3d}% | {bar:<10} | {advice}")
```

```python
import urllib.request
import json

# 1. 都市"
city = "kashiwa"
geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"
geo_res = urllib.request.urlopen(geo_url)
geo_data = json.loads(geo_res.read().decode("utf-8"))

lat = geo_data["results"][0]["latitude"]
lon = geo_data["results"][0]["longitude"]
name = geo_data["results"][0]["name"]
country = geo_data["results"][0]["country"]

# 2. 柏市の7日間の降水確率・最高/最低気温・平均湿度を取得する
weather_url = (
    f"https://api.open-meteo.com/v1/forecast?"
    f"latitude={lat}&longitude={lon}"
    f"&daily=precipitation_probability_max,temperature_2m_max,temperature_2m_min,relative_humidity_2m_mean"
    f"&timezone=Asia/Tokyo"
)
weather_res = urllib.request.urlopen(weather_url)
weather_data = json.loads(weather_res.read().decode("utf-8"))

dates = weather_data["daily"]["time"]
probs = weather_data["daily"]["precipitation_probability_max"]
temp_maxs = weather_data["daily"]["temperature_2m_max"]
temp_mins = weather_data["daily"]["temperature_2m_min"]
humids = weather_data["daily"]["relative_humidity_2m_mean"]

# 3. forループで1日ずつ見やすく表示する
print(f"📍 {name} ({country}) の1週間の天気予報（湿度・気温・降水確率）")
print("-" * 75)

for i in range(len(dates)):
    date = dates[i]
    prob = probs[i]
    t_max = temp_maxs[i]
    t_min = temp_mins[i]
    humid = humids[i]
    
    # None（データなし）対策
    if prob is None or t_max is None or t_min is None or humid is None:
        print(f"{date} : データなし")
        continue

    # 降水確率に応じて文字の棒（■）を作る（10%ごとに1個）
    bar = "■" * (prob // 10)
    
    # 確率に応じたアドバイス
    if prob >= 50:
        advice = "☔ 傘が必要"
    elif prob >= 30:
        advice = "🌂 折りたたみ傘"
    else:
        advice = "☀️ 傘不要"
        
    print(f"{date} | 湿度:{humid:3d}% | 気温:{t_min:4.1f}〜{t_max:4.1f}℃ | 降水:{prob:3d}% {bar:<10} | {advice}")
```
