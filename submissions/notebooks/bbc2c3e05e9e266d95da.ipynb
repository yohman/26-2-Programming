#　日本の総人口と65歳以上の比較
##　今井恭瑳

```python
import pandas as pd
```

```python
df = pd.read_csv('japan.csv')
print(df.shape)
```

```python
columns = ['都道府県', '総人口', '65歳以上人口', '合計特殊出生率']
data = df[columns]
data.head()
```

```python
top5 = df.sort_values('総人口', ascending=False).head(5)

for index, row in top5.iterrows():
    print(f"{row['都道府県']}: {row['総人口']:,} people")
```

```python
top5 = df.sort_values('65歳以上人口', ascending=False).head(5)

for index, row in top5.iterrows():
    print(f"{row['都道府県']}: {row['65歳以上人口']:,} people")
```

```python
top5 = df.tail(5)

for index, row in top5.iterrows():
    print(f"{row['都道府県']}: {row['総人口']:,} people")
```

```python
top5 = df.tail(5)

for index, row in top5.iterrows():
    print(f"{row['都道府県']}: {row['65歳以上人口']:,} people")
```

都道府県ごとの総人口と高齢者の人口は、上位５都道府県と下位５都道府県を比較した際にどちらもランキングに大きな差がなかったため、総人口と高齢者数はおおむね比例している。
