# My Chatbot｜ココロと話そう
言った言葉をまずオウム返しして、そのあと気分しだいで返事が変わる、お調子者のオウムのBot。

このBotはAIではなく、**自分で決めたルール**で返します。

## 1｜Botのプロフィール

````python
# 辞書（dict）：Botのプロフィール
bot = {
    "name": "ココロ",
    "role": "気分屋のオウム",
    "likes": ["ものまね", "パタパタ", "ひまわりの種"],  # リスト（list）
}

# 気分ごとの顔
faces = {
    "normal": "🦜",
    "happy": "🦜✨",
    "tired": "🦜💤",
}

print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
````

## 2｜三つの気分の返事
返事の先頭の `{echo}` は、あとで「入力した言葉のオウム返し」に置きかえる。

````python
import random

responses = {
    "normal": [
        "{echo} ふむふむ、それで？",
        "{echo} ココロだよ。よろしくピー。",
        f"{{echo}} ぼくは{bot['likes'][0]}が好き。きみは？",
        "{echo} …って言ってみたかっただけ！",
    ],
    "happy": [
        "{echo} パタパタ〜！サイコーだね！",
        "{echo} いいね！もう一回言って！",
        f"{{echo}} {bot['likes'][1]}したくなってきた！",
        "{echo} ホイホイ！今日はお祭りだピー！",
    ],
    "tired": [
        "{echo} …ねむ…ピ…",
        "{echo} …止まり木で、ちょっと休もう。",
        "{echo} …がんばりすぎたね。",
        f"{{echo}} …{bot['likes'][2]}の夢を見るピ…",
    ],
}

print(random.choice(responses["happy"]).replace("{echo}", "やったー！やったー！"))
````

## 3｜気分チェッカー
Cellを実行すると入力欄が出ます。「元気」「つかれた」「宇宙」などを入力しよう。
ココロは、入力した言葉を**二回くり返してから**返事をします。

````python
# 気分を判定する言葉のリスト
happy_words = ["元気", "楽しい", "うれしい", "最高", "happy", "good"]
tired_words = ["元気じゃない", "つかれ", "疲れ", "眠い", "つらい", "tired", "sad"]

question = "今日の気分はどう？ 元気？ つかれた？"
print(f"{faces['normal']} {bot['name']}：{question}")

try:
    message = input(f"{faces['normal']} {question}").strip()
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 判定用に小文字にそろえる（オウム返しには元の言葉を使う）
text = message.lower()

# まず normal にしておく
mood = "normal"

for word in happy_words:
    if word in text:
        mood = "happy"

# 「元気じゃない」は「元気」も含むので、tired を後に調べて上書きする
for word in tired_words:
    if word in text:
        mood = "tired"

# オウム返し：二回くり返す。入力が空なら「…？」
if message == "":
    echo = "…？"
else:
    echo = f"{message}！{message}！"

reply = random.choice(responses[mood]).replace("{echo}", echo)
print(f"{faces[mood]} {bot['name']}：{reply}")
````

## 自分のBotについて

**自分で追加したオリジナルの機能・特徴は？**
入力した言葉を二回くり返す「オウム返し」を、返事の先頭につけた。入力が空のときは「…？」と返す。三つの気分ごとに、返事を四つずつ自分で考えた。

**これからどんな機能を加え、どんなロボットに育てたい？**
話した言葉を覚えて、あとでふいに言い出す（本物のオウムのように）ようにしたい。気分が自分でころころ変わる、本当に気分屋のオウムに育てたい。
