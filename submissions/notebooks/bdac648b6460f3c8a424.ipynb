## 調べたいこと

札幌市の朝8時と夕方18時では、予報の相対湿度は何％違う？

出典：[Open-Meteo](https://open-meteo.com/en/docs)／取得日：2026年10月21日／時刻：日本時間

以下のURLは札幌市付近の湿度予報を返します。

````python
import json
from urllib.request import urlopen

# 札幌の緯度(43.06)・経度(141.35)と、相対湿度(relative_humidity_2m)を指定したURL
url = "https://api.open-meteo.com/v1/forecast?latitude=43.06&longitude=141.35&hourly=relative_humidity_2m&forecast_days=2&timezone=Asia%2FTokyo"

with urlopen(url, timeout=20) as response:
    data = json.load(response)
````

## 湿度を取り出す


0時が添字0なので、8時は`[8]`、18時は`[18]`です。

````python
humidity_list = data["hourly"]["relative_humidity_2m"]
morning = humidity_list[8]
evening = humidity_list[18]

print(f"対象日: {data['hourly']['time'][0][:10]}")
print(f"朝8時: {morning} %")
print(f"夕方18時: {evening} %")
print(f"朝と夕方の差: {morning - evening:.1f} %")
````

## 分かったこと

10月21日の予報では、札幌の朝8時の湿度は 92 ％、夕方18時は 94 ％。朝に比べて夕方の方が 2.0 ％ 高くなり、夕方にかけて少しジメジメする（湿度が上がる）ことが分かった。

**元データと照らして確かめた値：**
APIから取得したデータの `relative_humidity_2m` リストにおいて、8番目の要素が「92」、18番目の要素が「94」となっており、プログラムの出力値と目視で一致していることを確認した。

**このデータだけでは分からないこと（限界）：**
これはあくまで現在の予報データであり、実際の天候によって湿度は大きく変動するため、実測値とは異なる可能性がある。また、湿度だけでなく風速や気温の変化がどう影響しているかまではこのデータだけでは分からない。
