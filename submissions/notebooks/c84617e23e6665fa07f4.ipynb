## 1｜Botのプロフィール
**辞書**に名前・役割を、**リスト**に好きなものを三つ入れる。顔は気分ごとに別の辞書 `faces` で決める。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
# 辞書（dict）：Botのプロフィールを「名前: 値」の形で入れる
bot = {
    "name": "シェフ",             # Botの名前
    "role": "話し相手",  # Botの役割
    "likes": ["音楽", "ダイスゲーム"],  # リスト（list）：好きなもの三つ
    "adjective": ["焼き","生","蒸し","塩漬け","干し","冷えた","茹で","2日目の"],
    "food": ["ピータン","蕎麦","ブロック肉","八つ橋","芋","カレー","牡蠣","バラムツ","ビリヤニ","ふぐ"]
}

# 気分ごとの顔。キーは normal / happy / tired の三つ
faces = {
    "normal": "🤖",
    "happy": "🤖✨",
    "hungry": "🤖❍",
}

# プロフィールを表示する
print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を四つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random  # ランダムに選ぶ道具を使えるようにする

# 気分ごとの返事リスト（自分らしい言葉に変えよう）

favorite_songs= [
        "Again https://youtu.be/S2IOrt-n2fc?si=7JMoCHIgz7D3mXbQ",
        "キャンバス https://youtu.be/KSrGJKdQtt0?si=AYzg2sFDWeMp1uem",
        "ゴールデンタイムラバー https://youtu.be/0iAF8TJAqp4?si=lPAzFYfKQQ-V9gJC",]

user_dice = random.choice([1,2,3,4,5,6])
bot_dice = random.choice([1,2,3,4,5,6])
detayo=(f"あなたの目は {user_dice}")
detane=(f"{bot['name']}の目は {bot_dice}")
if user_dice > bot_dice :result=("あなたの勝ち！")
elif user_dice < bot_dice :result=(f"{bot['name']}の勝ち！")
else:result=("引き分け！")

responses = {
    "normal": [
        f"こんにちは！{bot['name']}です。",
        "その話、もう少し聞かせて。",
        f"私は{bot['likes'][0]}が好き。特に好きな曲は{random.choice(favorite_songs)}　です。あなたは？",
    ],
    "happy": [
        "その元気、少し分けてほしい！",
        "いいね！何があったの？",
        f"せっかくだし{bot['likes'][1]}をしよう！{detayo}！　 {detane}！　 {result}",
    ],
    "hungry": [
        f"シェフの気まぐれ{random.choice(bot['adjective'])}{random.choice(bot['food'])}ならあるよ！！おいしく食べてね！",
        f"{random.choice(bot['adjective'])}{random.choice(bot['food'])}ならあるよ！！おいしく食べてね！",
        f"{random.choice(bot['adjective'])}{random.choice(bot['food'])} と {random.choice(bot['adjective'])}{random.choice(bot['food'])} ならあるよ！！おいしく食べてね！",
    ],
}

# 確認：最初の返事と、ランダムな返事を表示
print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分チェッカー
**一回の実行で、一回返事をします。** 続けて話すときは、このCellをもう一度実行しよう。
「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。

言葉のリストに含まれる文字列を探す → `mood` を決める → その気分のリストから返す。

二つの `for` と `try / except` は準備済み。**今週はそのまま使ってOK。**
変えてみるのは、言葉のリスト・`if` の条件・返事です。

```python
# 1. 気分を判定する言葉のリスト（自分で増やしてみよう）
# 英語もOK（英語は小文字で書く）
happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "最高",
    "happy",
    "good",
    "great",
    "fun",
    "fine",
]
hungry_words = [
    "お腹空いた",
    "空腹",
    "ご飯",
    "食べたい",
    "hungry",
    "food",
]

# 2. Botが質問する
question = "今日の気分はどう？ 元気？ つかれた？ (How are you?)"
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える（入力が中断された場合は空欄として扱う）
try:
    message = input(f"{faces['normal']} {question}").strip().lower()  # lower() で小文字にそろえる
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

# ループは準備済み：happy の言葉があれば happy にする
for word in happy_words:
    if word in message:
        mood = "happy"

# 「元気じゃない」は「元気」も含むので、tired を後に調べて上書きする
for word in hungry_words:
    if word in message:
        mood = "hungry"

# 5. 気分に合う返事をランダムに選んで表示する
reply = random.choice(responses[mood])
print(f"{faces[mood]} {bot['name']}：{reply}")
```

> **Aha!** 「元気じゃない」には「元気」も含まれる。最初のループで `happy` になり、後のループで `tired` に上書きされる。二つのループの順番を入れ替えたら、どちらが残る？
