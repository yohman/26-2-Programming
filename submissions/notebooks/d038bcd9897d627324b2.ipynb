## 調べたいこと

柏の今後7日間の予報で、最高気温・最低気温・寒暖差が一番大きい日はいつ？

出典：[Open-Meteo](https://open-meteo.com/en/docs)／取得日：2026年10月21日／時刻：日本時間

```python
import json
from urllib.request import urlopen

url = "https://api.open-meteo.com/v1/forecast?latitude=35.86&longitude=139.97&daily=temperature_2m_max,temperature_2m_min&forecast_days=7&timezone=Asia%2FTokyo"

try:
    with urlopen(url, timeout=20) as response:
        data = json.load(response)
    # 提出用に保存しておく
    with open("weather-kashiwa-week.json", "w", encoding="utf-8") as file:
        json.dump(data, file, ensure_ascii=False)
except Exception as error:
    print("通信できないので保存したJSONを使います：", error)
    with open("weather-kashiwa-week.json", encoding="utf-8") as file:
        data = json.load(file)

print(list(data.keys()))
```

## 気温を取り出す

`data["daily"]`の中に、日付・最高気温・最低気温のリストがあります。同じ添字が同じ日です。

```python
daily = data["daily"]
dates = daily["time"]
highs = daily["temperature_2m_max"]
lows = daily["temperature_2m_min"]

for date, high, low in zip(dates, highs, lows):
    print(f"{date}  最高 {high} ℃  最低 {low} ℃  差 {high - low:.1f} ℃")
```

## 一番を探す

```python
diffs = [high - low for high, low in zip(highs, lows)]
i = diffs.index(max(diffs))

print(f"寒暖差が一番大きい日：{dates[i]}")
print(f"最高 {highs[i]} ℃ / 最低 {lows[i]} ℃ / 差 {diffs[i]:.1f} ℃")

# 元データと照らして確認
print("確認：元JSONの値", data["daily"]["temperature_2m_max"][i], data["daily"]["temperature_2m_min"][i])
```

## 分かったこと

10月21日が寒暖差が一番大きく、最高27.4℃・最低18.2℃で差は9.2℃だった。

**確認：** 元のJSONの`temperature_2m_max`と`temperature_2m_min`の添字0の値（27.4と18.2）と、表示した値が同じことを確かめた。

**このデータだけでは分からないこと：** これは予報で実測ではない。風・雨・湿度は調べていない。取得日が変わると値も変わる。
