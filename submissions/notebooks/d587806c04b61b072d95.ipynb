```python
import pandas as pd  # ← この1行を追加してください！

# 表用データの作成
table_data = []
for name, info in players_data.items():
    table_data.append(
        {
            "棋士名": name,
            "段位・称号": f"{info['title']}（{info['rank']}）",
            "通算成績": f"{info['wins']}勝 {info['losses']}敗",
            "勝率": f"{info['win_rate']*100:.1f}%",
            "得意戦法": info["style"],
        }
    )

df = pd.DataFrame(table_data)
display(df)
```

```python
import ipywidgets as widgets 
from IPython.display import HTML, display  


# カード表示用の関数
def show_player_card(player_name):
    p = players_data[player_name]
    win_rate_pct = f"{p['win_rate']*100:.1f}%"

    card_html = f"""
    <div style="border: 1px solid #e0e0e0; border-radius: 12px; padding: 20px; max-width: 500px; background-color: #ffffff; box-shadow: 0 4px 6px rgba(0,0,0,0.05); font-family: sans-serif;">
        <div style="display: flex; align-items: center; gap: 20px;">
            <div style="flex-shrink: 0;">
                <img src="{p['img']}" alt="{player_name}" style="width: 100px; height: 125px; object-fit: cover; border-radius: 8px; background-color: #f3f4f6; border: 1px solid #e5e7eb;">
            </div>
            <div>
                <h2 style="margin: 0; font-size: 22px; color: #111;">{player_name} <span style="font-size: 14px; color: #666;">{p['rank']}</span></h2>
                <p style="margin: 4px 0 12px 0; color: #4f46e5; font-weight: bold; font-size: 14px;">{p['title']}</p>
                <p style="margin: 2px 0; font-size: 13px; color: #555;"><b>生年月日:</b> {p['birth']}</p>
                <p style="margin: 2px 0; font-size: 13px; color: #555;"><b>得意戦法:</b> {p['style']}</p>
            </div>
        </div>
        <hr style="border: none; border-top: 1px solid #f0f0f0; margin: 16px 0;">
        <div style="display: flex; justify-content: space-around; text-align: center;">
            <div>
                <span style="font-size: 12px; color: #777;">通算勝率</span>
                <p style="margin: 4px 0 0 0; font-size: 18px; font-weight: bold; color: #059669;">{win_rate_pct}</p>
            </div>
            <div>
                <span style="font-size: 12px; color: #777;">通算対戦成績</span>
                <p style="margin: 4px 0 0 0; font-size: 18px; font-weight: bold; color: #1f2937;">{p['wins']}勝 {p['losses']}敗</p>
            </div>
        </div>
        <p style="margin-top: 16px; font-size: 12px; color: #666; line-height: 1.5; background: #f9fafb; padding: 10px; border-radius: 6px;">
            {p['description']}
        </p>
        <div style="margin-top: 12px; text-align: right;">
            <a href="{p['url']}" target="_blank" style="display: inline-block; font-size: 12px; color: #4f46e5; font-weight: bold; text-decoration: none; border: 1px solid #4f46e5; padding: 6px 12px; border-radius: 6px; transition: 0.2s;">
                公式プロフィールを見る ➔
            </a>
        </div>
    </div>
    """
    display(HTML(card_html))


# ドロップダウンUIの作成と表示
dropdown = widgets.Dropdown(
    options=list(players_data.keys()),
    value="藤井聡太",
    description="棋士選択:",
)

widgets.interact(show_player_card, player_name=dropdown)
```
