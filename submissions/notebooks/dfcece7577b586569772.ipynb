## 調べたいこと

那覇の朝8時と夜22時では、予報の湿度何％違う？

出典：[Open-Meteo](https://open-meteo.com/en/docs)／取得日：2026年10月20日／時刻：日本時間

以下のURLは柏付近の気温予報を返します。追加インストールは不要です。

````python
import json
from urllib.request import urlopen

url = "https://api.open-meteo.com/v1/forecast?latitude=35.86&longitude=139.97&hourly=temperature_2m&forecast"

with urlopen(url, timeout=18) as response:
    data = json.load(response)
````

## 湿度を取り出す

`data`は辞書。その中の`hourly`に、時刻と気温のリストがあります。

0時が添字0なので、8時は`[8]`、22時は`[22]`です。

````python
import json
from urllib.request import urlopen

url = "https://api.open-meteo.com/v1/forecast?latitude=26.21&longitude=127.68&hourly=relative_humidity_2m"

with urlopen(url, timeout=18) as response:
    data = json.load(response)

humidities = data["hourly"]["relative_humidity_2m"]
morning = humidities[8]    
night = humidities[22]     

print(f"対象日: {data['hourly']['time'][0][:10]}")
print(f"朝8時: {morning}%")
print(f"夜22時: {night}%")
print(f"夜と朝の差: {abs(night - morning)}%")
````

## 分かったこと

10月15日の予報では、朝8時は74％、夜22時は88％。夜の方が14％高かった。

**確認：** 元のJSONの`temperature_2m`で、添字8が17.7、添字14が21.1であることを確かめた。

**限界：** これは予報で、実測ではない。風や雨は調べていない。取得日が変わると値も変わるので、再実行したら観察も書き直す。
