## トップ5をとる



```python
import json
from urllib.request import urlopen

# 日本のApple Musicで最も再生されている曲トップ5のJSONデータを取得するURL
url = "https://rss.applemarketingtools.com/api/v2/jp/music/most-played/5/songs.json"

# データの取得
with urlopen(url, timeout=20) as response:
    data = json.load(response)

# トップ5の曲情報をループ処理で綺麗に表示
print("🎵 Apple Music 日本のトップソング 5 🎵\n")
for i, song in enumerate(data["feed"]["results"], start=1):
    song_name = song["name"]          # 曲名
    artist_name = song["artistName"]  # アーティスト名
    
    print(f"{i}位: {song_name} / {artist_name}")
```
