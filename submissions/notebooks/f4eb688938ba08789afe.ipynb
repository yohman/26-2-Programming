## 何ができるNotebookか
札幌（北海道）・那覇（沖縄）・柏（千葉）の「明日」の天気予報（最高気温・最低気温・降水確率）を自動で取得し、3都市を並べて表示します。さらに札幌と那覇の最高気温の差を計算し、差の大きさに応じたコメントを表示します。

## 天気データの出典
- Open-Meteo Weather Forecast API：https://open-meteo.com/
- API仕様（使用エンドポイント）：https://api.open-meteo.com/v1/forecast

## 自分たちで追加した機能・特徴
- 比較する都市に、札幌・那覇に加えて**柏**を追加した
- 降水確率を `█`（10%ごとに1つ）の**棒グラフ風**に表示した
- 札幌と那覇の**最高気温の差**を計算して表示した
- 気温差が基準（`diff_line`）以上かどうかで**コメントを切り替える**ようにした
- 気温差が小さいときは**アスキーアート**を表示する遊び心を入れた
- 降水確率が取得できない場合（`None`）に「データなし」と表示する**エラー対策**を入れた

## これから改善したいこと・追加したい機能
- 都市を自由に追加・選択できるようにする（入力で都市名を指定）
- 設定だけで使っていない `rain_line`（傘の基準）を使い、「傘が必要」などの**服装・持ち物アドバイス**を表示する
- 気温や降水確率を**グラフ（matplotlib）**で見やすく表示する
- 都市名の表記を直す（`Kasiwa` → `Kashiwa`）
- 通信エラー時にメッセージを出して止まらないようにする
- 週間予報の表示や、天気アイコンの追加

```python
import json
from urllib.request import urlopen
from urllib.parse import urlencode

# 都市名 → 緯度経度の辞書
cities = {
    "Sapporo": {"lat": 43.06, "lon": 141.35},
    "Naha":    {"lat": 26.21, "lon": 127.68},
    "Kasiwa": {"lat":35.87,"lon":139.98}
}

past_days = 1       # 先頭(0番目)が昨日
forecast_days = 3   # 1番目=今日、2番目=明日
tomorrow = 2        # 明日の位置(添字)
daily_fields = "temperature_2m_max,temperature_2m_min,precipitation_probability_max"

results = {}

for city, place in cities.items():
    url = "https://api.open-meteo.com/v1/forecast?" + urlencode({
        "latitude": place["lat"], "longitude": place["lon"],
        "daily": daily_fields,
        "past_days": past_days, "forecast_days": forecast_days,
        "timezone": "auto",
    })
    with urlopen(url, timeout=10) as response:
        weather = json.load(response)

    d = weather["daily"]
    results[city] = {
        "date": d["time"][tomorrow],
        "max": d["temperature_2m_max"][tomorrow],
        "min": d["temperature_2m_min"][tomorrow],
        "rain": d["precipitation_probability_max"][tomorrow],
    }

results
```

```python
for city, r in results.items():
    rain = r["rain"]
    if rain is None:
        bar, rain_text = "データなし", "データなし"
    else:
        bar = "█" * (rain // 10)
        rain_text = f"{rain}%"
    print(f'{city:8} {r["date"]}  最高{r["max"]}℃ 最低{r["min"]}℃  降水確率 {bar} {rain_text}')

# --- ここからチームで決める ---
diff = results["Sapporo"]["max"] - results["Naha"]["max"]
print(f"\n最高気温の差: {diff:.1f}℃")

diff_line = 20   # 例: 何℃以上なら「服装を大きく変える」にする?
rain_line = 40   # 例: 何%以上なら「傘がいりそう」にする?
art = r"""
　　　　　～～～～～～～～、
　　　　/｜ｏ　〉┴〈_〉┴〈　0｜
　　　/　|　　｜◎｜◎　／）｜
　　　|ｏ　|　ｒ ヽ、_人 ／／ O｜
　　　｜｜　ヽ＼＿＿＿_フ　 |
　　　| Ｏ｜Ｏ ＼　　凵 凵　0 |　　　　/)
　　　｜0 |　　０ ＼⌒⌒V O｜　　／っ）
　　　/T　|　o　0　　｀ー_´ｏ　｜／／
　　 |_」|／|￣￣￣▽▼▽￣￣|
　　 | | |__｜＿＿＿　■＿＿＿|
　　｜| | =|＝＝＝＝▼＝＝＝|
"""

if abs(diff) >= diff_line:
    print("感想:最高気温の差、ジェットコースターかよｗｗｗ(ノಠ益ಠ)ノ")
else:
    print(art)
    print("感想:地球働け( ՞ਊ ՞)")
```
