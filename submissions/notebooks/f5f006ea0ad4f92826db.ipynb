# ポケモン4匹の能力を比較

**知りたいこと：攻撃が高いポケモンは、素早さも高いのか？**

PokeAPIのデータを保存したJSONから、4匹の「攻撃」と「素早さ」を比べます。

```python
import json
from pathlib import Path

with open('pokemon_api_saved.json', encoding='utf-8') as f:
    pokemon = json.load(f)  # リストの中に辞書が入っている
print('読み込んだ件数:', len(pokemon))
print('最初のデータ:', pokemon[0])
```

```python
print('名前       攻撃  素早さ  差(素早さ-攻撃)')
for p in pokemon:
    s = p['stats']
    print(f"{p['name']:<7} {s['attack']:>4} {s['speed']:>5} {s['speed']-s['attack']:>8}")

fastest = max(pokemon, key=lambda p:p['stats']['speed'])
strongest = max(pokemon, key=lambda p:p['stats']['attack'])
print('\n素早さが最も高い:', fastest['name'], fastest['stats']['speed'])
print('攻撃が最も高い:', strongest['name'], strongest['stats']['attack'])
```

```python
import matplotlib.pyplot as plt
names=[p['name'] for p in pokemon]
# 日本語フォントのない環境でも文字化けしないよう、図ではAPIの英語名を使用
labels=[p['api_name'] for p in pokemon]
attack=[p['stats']['attack'] for p in pokemon]
speed=[p['stats']['speed'] for p in pokemon]
x=list(range(len(pokemon)))
plt.figure(figsize=(8,4))
plt.bar([i-0.18 for i in x],attack,width=0.36,label='Attack')
plt.bar([i+0.18 for i in x],speed,width=0.36,label='Speed')
plt.xticks(x,labels)
plt.ylabel('Base stat'); plt.title('Attack vs Speed (PokeAPI)'); plt.legend(); plt.tight_layout(); plt.show()
```

## 分かったこと

- **結果：** ドラパルトは攻撃120、素早さ142で、どちらも4匹中1位でした。一方、ハピナスは攻撃10、素早さ55でした。4匹の中では攻撃と素早さの両方が高い例もありますが、必ず同じ傾向になるとは言えません。
- **元データの確認：** 保存JSONのドラパルトの `attack` は120、`speed` は142です。表示結果と一致しました。
- **分からないこと：** この4匹だけでは、全ポケモンにも同じ傾向があるかは分かりません。また、実際の対戦の強さは技やレベルなどにも左右されます。
