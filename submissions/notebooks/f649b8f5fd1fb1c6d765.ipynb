## 調べたいこと

keshiの楽曲の長さにはどのような違いがあるのか。平均の長さや最も長い曲・短い曲を比較して調べる。

出典：　データ提供元：Apple iTunes Search API
　　　　データ取得先：https://itunes.apple.com/search?term=keshi&entity=song&country=jp&limit=50
　　　　データ取得日：2026年10月21日


```python

import json
from urllib.request import urlopen
from urllib.parse import urlencode

# keshiの楽曲を検索するURLを作る
params = {
    "term": "keshi",
    "entity": "song",
    "country": "jp",
    "limit": 50
}

url = "https://itunes.apple.com/search?" + urlencode(params)

# APIからデータを取得する
with urlopen(url, timeout=20) as response:
    data = json.load(response)

print("データを取得しました")
print("検索結果の件数：", data["resultCount"])
```

```python

songs = []
seen = set()

for item in data["results"]:
    artist = item.get("artistName", "")
    track = item.get("trackName", "")

    if artist in ["keshi", "ケシ"] and "trackTimeMillis" in item:
        # 曲名と再生時間の組み合わせで重複を確認
        key = (track.strip().lower(), item["trackTimeMillis"])

        if key in seen:
            continue

        seen.add(key)

        songs.append({
            "track": track,
            "artist": artist,
            "length_seconds": item["trackTimeMillis"] / 1000
        })

print(f"重複を除いた曲数: {len(songs)}")

for song in songs:
    minutes = int(song["length_seconds"] // 60)
    seconds = int(song["length_seconds"] % 60)
    print(f'{song["track"]} - {minutes}:{seconds:02d}')
```

```python

# 楽曲の長さを比較する
if len(songs) > 0:
    lengths = [song["length_seconds"] for song in songs]

    average = sum(lengths) / len(lengths)
    longest = max(songs, key=lambda song: song["length_seconds"])
    shortest = min(songs, key=lambda song: song["length_seconds"])

    print(f"重複を除いた曲数: {len(songs)}")
    print(f"平均の長さ: {int(average // 60)}分{int(average % 60):02d}秒")

    print(
        f"最も長い曲: {longest['track']} "
        f"({int(longest['length_seconds'] // 60)}分"
        f"{int(longest['length_seconds'] % 60):02d}秒)"
    )

    print(
        f"最も短い曲: {shortest['track']} "
        f"({int(shortest['length_seconds'] // 60)}分"
        f"{int(shortest['length_seconds'] % 60):02d}秒)"
    )
else:
    print("該当する楽曲が見つかりませんでした。")
```

## 分かったこと

取得した楽曲の平均の長さは3分05秒だった。最も長い曲は「us」で4分08秒、最も短い曲は「Amen」で1分01秒だった。このことから、今回取得した楽曲の中でも、曲によって長さに大きな違いがあることが分かった。今回の検索結果は最大50件に限られているため、keshiのすべての楽曲を調べられたとは限らない。また、曲の長さだけでは人気や歌詞の内容までは分からない。
