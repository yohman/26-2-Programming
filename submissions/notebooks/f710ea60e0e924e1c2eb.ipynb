# My Chatbot｜キャラクターと話そう
普段の自分とお喋りできるBOT

```python

bot = {
    "name": "俺bot",             
    "role": "普段の自分", 
    "likes": ["ゲーム", "アニメ", "お金"], 
}

faces = {
    "normal": "ゲーム、ゲーム🤖",
    "happy": "最高にハイってやつだ🤖✨",
    "tired": "今日は疲れた🤖💤",
}

print(f"{faces['normal']} {bot['name']}｜{bot['role']}")
```

```python
import random  

responses = {
    "normal": [
        f"こんにちは！{bot['name']}です。",
        "ゲームしようぜ。",
        f"俺は{bot['likes'][0]}が好き。あなたは？",
        "最近、ゲームできてる？",
    ],
    "happy": [
        "最高に清々しい気分だ",
        "いいね！何があったの？",
        "うれしいことは、誰かに話したくなるね。",
        f"{bot['likes'][1]}観ようぜ",
    ],
    "tired": [
        "おつかれー",
        "もう眠いぞ",
        f"{bot['likes'][2]}もないし",
    ],
}

print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

```python

happy_words = [
    "元気",
    "楽しい",
    "うれしい",
    "嬉しい",
    "最高にハイってやつだ",

]
tired_words = [
    "元気じゃない",
    "つかれた",
    "眠い",
    "しんど",
    "つらい",

]

# 2. Botが質問する
question = "今日の調子はどう？ "
print(f"{faces['normal']} {bot['name']}：{question}")

# 3. ユーザーが答える（入力が中断された場合は空欄として扱う）
try:
    message = input(f"{faces['normal']} {question}").strip().lower()  # lower() で小文字にそろえる
except KeyboardInterrupt:
    message = ""
print(f"👤 あなた：{message}")

# 4. 気分を決める（まず normal にしておく）
mood = "normal"

# ループは準備済み：happy の言葉があれば happy にする
for word in happy_words:
    if word in message:
        mood = "happy"

# 「元気じゃない」は「元気」も含むので、tired を後に調べて上書きする
for word in tired_words:
    if word in message:
        mood = "tired"

# 5. 気分に合う返事をランダムに選んで表示する
reply = random.choice(responses[mood])
print(f"{faces[mood]} {bot['name']}：{reply}")
```
