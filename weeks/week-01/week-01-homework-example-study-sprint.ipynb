## 何をつくる？
勉強を始める前に、集中時間と休憩時間を合わせて、全部で何分必要かを知らせる小さなツールです。

## 設計メモ / Human Coding

- **ツールの名前：** 学習スプリントプランナー
- **だれの用事を助ける？** 勉強を始める前に、必要な時間を知りたい人。
- **入力：** 名前、取り組むこと、1回の集中時間、休憩時間、集中する回数。
- **変数：** `focus_minutes`、`break_minutes`、`session_count` など。
- **計算：** 集中の合計と、途中の休憩の合計を足す。
- **最後に伝えること：** 何に取り組み、全部で何分確保すればよいか。

## 動かしてみる
下のセルを実行し、表示される質問に答えます。この例の保存済み出力は「あおい／英語の単語／25分／5分／3回」で実行したものです。

```python
# 勉強する人と、その日のテーマを聞く
student_name = input('名前: ')
task = input('取り組むこと: ')

# input()は文字列を返すので、計算する値は整数に変える
focus_minutes = int(input('1回の集中時間（分）: '))
break_minutes = int(input('休憩時間（分）: '))
session_count = int(input('集中する回数: '))

total_focus = focus_minutes * session_count
# 最後の集中の後は休まないので、休憩は集中より1回少ない
total_break = break_minutes * (session_count - 1)
total_minutes = total_focus + total_break

print('━━ 学習スプリントプラン ━━')
print(f'{student_name}さんのテーマ: {task}')
print(f'{focus_minutes}分集中 × {session_count}回')
print(f'集中 {total_focus}分 + 休憩 {total_break}分')
print(f'合計 {total_minutes}分を確保しよう。')
```

## Verification / 二通りで確認

| 入力 | 予想 | 実際 | 気づいて直したこと |
| --- | --- | --- | --- |
| あおい／英語の単語／集中25分／休憩5分／3回 | 集中75分＋休憩10分＝85分 | 85分 | 最初は休憩を3回にして90分とした。最後の回の後には休憩がないので、`session_count - 1` に直した。 |
| りん／数学の復習／集中40分／休憩10分／2回 | 集中80分＋休憩10分＝90分 | 90分 | 別の入力でも計算と表示が変わることを確認した。 |

**観察：** `input()` で受けた `25` は文字列なので、`int()` を使わないと数値として合計できない。

出力を読みやすくする一文だけをAIに相談した。`合計85分` だけの短い表示を提案されたが、**修正して採用**した。集中75分と休憩10分を別々に出したほうが、合計85分の根拠を自分でも確認できるからです。二つの入力で出力を確かめました。
