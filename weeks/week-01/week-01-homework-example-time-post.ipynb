## 1｜設計メモ

- **未来の自分は？** 月面図書館の夜間便で働く私。これは予言ではなく、手紙を楽しくするためのフィクション。
- **入力：** 名前、挑戦したいこと、何年後か。
- **計算：** `2026 + years_ahead` で消印の年を決める。
- **口調：** 少しユーモラス。でも、今すぐできる小さな行動を一つ伝える。
- **変わる部分：** 年、名前、目標、経過年数。

## 2｜動く手紙

次のセルを実行し、三つの質問に答えます。入力は `input()` で受け取り、年数だけ `int()` で数に変えます。

```python
# 手紙に入れる言葉と、何年後から届くかを聞く
name = input('名前: ')
goal = input('今、挑戦したいこと: ')
years_ahead = int(input('何年後の自分から？（数字）: '))

# input()の答えは文字列なので、数に変えた年数を2026に足す
future_year = 2026 + years_ahead

# 名前・目標・年を手紙の中に入れる
print('📮 時空郵便局 ── 未来便')
print(f'消印: {future_year}年')
print(f'差出人: {future_year}年の{name}（月面図書館・夜間便）')
print()
print(f'{name}へ')
print(f'「{goal}」と書いたあの日のメモ、まだ持っているよ。')
print('最初の試作品は、飛ばずに坂を下っただけ。少し笑ったけれど、失敗の理由をノートに残した。')
print(f'{years_ahead}年後の私は、そのノートを次の設計図にしている。')
print('今日は10分だけ、作りたいものの絵を描いてみて。未来はそこから始まる。')
```
