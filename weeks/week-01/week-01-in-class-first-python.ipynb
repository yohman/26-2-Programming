## 0. プログラムは何をするのか

プログラムには、**入力（input）→ 命令（instructions）→ 出力（output）**があります。Pythonは上から下へ、一行ずつ命令を実行します。セルを選び、Run を押すか `Shift + Enter` を押して実行しましょう。

## 1. 最初の命令：`print()`

`print()` は、かっこの中の値を出力する関数です。実行前に、三行がそれぞれ何を表示するかノートに予想してください。

```python
print('Hello, world!')
print(10 + 7)
print('10 + 7')
```

### 観察

引用符 `' '` または `" "` で囲んだものは**文字列**（text）です。引用符のない `10 + 7` は、Pythonが計算する**式**です。

`#` の後ろは**コメント**です。人に向けたメモであり、Pythonは実行しません。

```python
# この行はPythonに実行してほしい命令です
print(3 * 4)  # かけ算の結果を表示します
print('3 * 4')  # こちらは文字として表示します
```

## 4. 変数：値に名前を付けて覚える

`=` は数学の「等しい」ではありません。右側を計算し、その値を左側の名前に**保存する**命令です。変数名は意味が分かるように、小文字と `_` を使って書きます。

```python
ticket_price = 1900
drink_price = 350
people = 2
total_cost = (ticket_price + drink_price) * people
print(total_cost)
```

### 変数の状態が変わる

次のセルを一行ずつ実行し、`score` が何を覚えているか追跡してください。`score = score + 5` は、以前の値を使って新しい値を作り、同じ名前に保存し直します。

```python
score = 10
score = score + 5
score += 2
print(score)
```

## 5. 値には型がある

Pythonは値の種類（**型**）を区別します。`str` は文字列、`int` は整数、`float` は小数、`bool` は真偽値です。`type()` で確かめられます。

```python
name = 'Rina'
year = 2026
temperature = 26.5
is_raining = False

print(type(name))
print(type(year))
print(type(temperature))
print(type(is_raining))
```

### 文字と数は同じようには足せない

次のセルはエラーになります。エラーメッセージの最後の一行を読み、何と何を足そうとしたのか説明してください。その後、`int()` を使って直します。

```python
age_as_text = '18'
# print(age_as_text + 1)  # 予想してから、必要ならコメントを外して試す
age = int(age_as_text)
print(age + 1)
```

## 6. 入力を受け取り、結果を伝える

`input()` はキーボードから受け取ったものを**文字列**として返します。計算に使うときは `int()` または `float()` で変換します。f文字列（`f'...'`）では、`{ }` の中に変数を入れられます。

```python
name = input('名前を入力してください: ')
minutes = int(input('今日プログラミングを練習する予定の分数: '))
hours = minutes / 60
print(f'{name}さんは、約{hours}時間プログラミングを練習する予定です。')
```
