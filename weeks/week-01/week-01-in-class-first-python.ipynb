## 0. プログラムは何をするのか

プログラムには、**入力（input）→ 命令（instructions）→ 出力（output）**があります。Pythonは上から下へ、一行ずつ命令を実行します。セルを選び、Run を押すか `Shift + Enter` を押して実行しましょう。

## 1. 最初の命令：`print()`

`print()` は、かっこの中の値を出力する関数です。実行前に、三行がそれぞれ何を表示するかノートに予想してください。

```python
print('Hello, world!')
print(10 + 7)
print('10 + 7')
```

### 観察

引用符 `' '` または `" "` で囲んだものは**文字列**（text）です。引用符のない `10 + 7` は、Pythonが計算する**式**です。

`#` の後ろは**コメント**です。人に向けたメモであり、Pythonは実行しません。

```python
# この行はPythonに実行してほしい命令です
print(3 * 4)  # かけ算の結果を表示します
print('3 * 4')  # こちらは文字として表示します
```

## 5. 値には型がある

Pythonは値の種類（**型**）を区別します。`str` は文字列、`int` は整数、`float` は小数、`bool` は真偽値です。`type()` で確かめられます。

```python
name = 'Rina'
year = 2026
temperature = 26.5
is_raining = False

print(type(name))
print(type(year))
print(type(temperature))
print(type(is_raining))
```

## 6. 入力を受け取り、結果を伝える

`input()` はキーボードから受け取ったものを**文字列**として返します。計算に使うときは `int()` または `float()` で変換します。f文字列（`f'...'`）では、`{ }` の中に変数を入れられます。

```python
name = input('名前を入力してください: ')
minutes = int(input('今日プログラミングを練習する予定の分数: '))
hours = minutes / 60
print(f'{name}さんは、約{hours}時間プログラミングを練習する予定です。')
```
