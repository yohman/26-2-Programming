```python
import folium

# Create a base map
m = folium.Map(location=[20, 0], zoom_start=2)

# Normalize the magnitudes to the range [2, 40]
min_radius = 2
max_radius = 40
min_mag = df['mag'].min()
max_mag = df['mag'].max()

def normalize_radius(mag, min_mag, max_mag, min_radius, max_radius):
    return ((mag - min_mag) / (max_mag - min_mag)) * (max_radius - min_radius) + min_radius

# Add circles to the map
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=normalize_radius(row['mag'], min_mag, max_mag, min_radius, max_radius),
        color='blue',
        fill=True,
        fill_color='blue'
    ).add_to(m)

# Display the map
m
```
