## 1.2 変数 Variables

プログラムには「変数」という仕組みがあります。プログラムを使うと便利な理由の1つです。

Variables is a fundamental part of coding. Let's find out why it is useful. Consider the following code:

````python
print("掛け算九九における2の段の答えは")
print(2 * 1)
print(2 * 2)
print(2 * 3)
print(2 * 4)
print(2 * 5)
print(2 * 6)
print(2 * 7)
print(2 * 8)
print(2 * 9)
````

```
変数 = 値
```
のようにプログラムを書き、「変数に値を代入(保存）する」という意味になります。私たちがこれまで慣れ親しんできた数式では、「変数と値は等しい」という意味でしたが、プログラムでは「=」は代入を意味します。（ちなみに「等しい」は「==」）

Changing the ```2``` to a ```5``` would be quite a time consuming task. This is where using a variable comes in handy.

### 1.2.3 文字列の中に数値を埋め込む
```format```　関数は、Pythonの文字列フォーマットを行うための強力なツールです。この関数を使用すると、文字列内にプレースホルダーを配置し、そこに値を挿入することができます。以下にformat関数の主な特徴と使用方法を説明します:

````python
radius = 5  # 半径 radius
print('半径が　5の円の円周の長さは')
print(radius * 2 * 3.14)
````

````python
radius =15000
print('半径が{}の円の円周の長さは'.format(radius))
print(radius * 2 * 3.14)
````

````python
#変数の値を複数個所に埋め込みたい場合は、次のように埋め込みたい個所に{}を、format関数の中には出力したい順番に変数を,区切りで入れます。
age = 25
print('今の年齢は{}歳です。来年は{}歳になります。'.format(age,age+1))
````

#### f-string

```f-string```（フォーマット済み文字列リテラル）は、Python 3.6以降で導入された文字列フォーマットの新しい方法です。```f-string```は、format()メソッドよりも簡潔で読みやすく、多くの場合でより高速です。

##### 基本的な使い方
```f-string```は、文字列の前にfまたはFを付けて作成します。中括弧{}内に直接Pythonの式を書くことができます。

````python
radius = 500
print(f'半径が{radius}の円の円周の長さは')
print(radius * 2 * 3.14)
````

````python
#変数の値を複数個所に埋め込みたい場合は、次のように埋め込みたい個所に{変数名}を入れてあげれば良いです。
age = 41
print(f'今の年齢は{age}歳です。来年は{age+1}歳になります。さ来年は{age+2}歳になります。')
````

たとえば冒頭の掛け算のプログラムは、2行目を次のように書き換えることで、1行目の変数のところだけを書き換えればよくなります。

````python
x = 13 #xという変数に、5という数値を代入
print(f'掛け算九九における{x}の段の答えは')
print(x * 1) #→5 × 1
print(x * 2) #→5 × 2
print(x * 3) #→5 × 3
print(x * 4) #→5 × 4
print(x * 5) #→5 × 5
print(x * 6) #→5 × 6
print(x * 7) #→5 × 7
print(x * 8) #→5 × 8
print(x * 9) #→5 × 9
````
