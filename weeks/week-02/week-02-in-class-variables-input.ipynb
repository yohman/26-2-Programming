## 1. `print()`、引用符、コメント
文字列は`' '`または`" "`で囲みます。`#`以降は人のためのコメントで、Pythonは実行しません。

```python
# 数値を出力する
print(19999)
print(10)

# 文字列を出力する
print('Hello World!')
print("今日はPythonを学ぶ")

# この行はコメント
# print('この行は実行されない')
```

### 確認 1
`print(10 + 7)` と `print('10 + 7')` の出力を予想し、実行後に違いを一文で説明してください。

```python
print(10 + 7)
print('10 + 7')
```

## 2. 式と演算
Pythonは計算が得意です。`+ - * /`に加えて、`%`は割った余りを求めます。演算の結果にも型があります。

```python
print(1 + 7)
print(1 - 7)
print(1 * 7)
print(1 / 7)
print(17 % 5)
print(type(1 / 7))
```

## 3. 変数：値に名前を付ける
`=`は「右の値を左の名前に入れる」代入です。数学の等号ではありません。変数名には意味のある英数字と`_`を使い、先頭に数字は使えません。

```python
student_name = 'Rina'
score = 80
print(student_name)
print(score)

# まとめて代入
a, b = 12, 5
print(a + b)
```

## 4. 自分自身への代入
`score = score + 5`は、古い`score`に5を足し、新しい値を同じ名前に保存します。`+=`はその短縮形です。各行の後の`score`を追跡してください。

```python
score = 70
score = score + 5
score += 10
print(score)
```

## 5. データ型と型変換
`input()`の結果は文字列（`str`）です。数値計算には`int()`または`float()`で変換します。変換できない文字列ではエラーになります。

```python
text_number = '12'
print(type(text_number))
number = int(text_number)
print(number + 3)
print(type(number))

# 実行したい場合は次の行の # を外す
# age = int(input('年齢を入力: '))
# print(age + 1)
```

## 6. f-stringで読みやすく出力する
文字列の先頭に`f`を付け、`{}`の中に変数や式を書きます。出力は人が読める文章にします。

```python
name = 'Rina'
price = 1200
tax_rate = 0.10
total = price * (1 + tax_rate)
print(f'{name}さんの税込み価格は ¥{total:.0f} です。')
```
