## 今日の目標
リストは順序のある集まり、辞書はキーと値の集まりです。情報を構造化し、変更をコミットとして残します。

## 1. リスト
添字は0から始まります。`append()`は末尾に一つの項目を追加します。

```python
scores = [80, 96, 90]
print(scores[0])
print(sum(scores) / len(scores))
scores.append(100)
print(scores)
```

## 2. 辞書
辞書は名前付きの値を扱います。存在しないキーを読むとエラーになります。

```python
profile = {'name': 'Yuki', 'year': 1, 'links': ['https://example.com']}
print(profile['name'])
profile['year'] = 2
print(profile)
```
