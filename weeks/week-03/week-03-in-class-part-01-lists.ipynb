## 2｜リストは、順番のある集まり

`[ ]`の中に、カンマで区切って値を置きます。`sum()`は合計、`len()`は個数。科目が増えても同じ計算式を使えます。`len`はlengthの略。


```python
scores = [80, 96, 90]
print(scores)
print(type(scores))
print(f"合計点：{sum(scores)}")
print(f"平均点：{sum(scores) / len(scores)}")
```

## 5｜スライスは「ここから、ここまでの手前」

`a[0:3]`は0・1・2番目。終了位置の3番目は含みません。省略すると先頭／末尾まで。まず下の四つの出力を予想しよう。


```python
a = [0, 10, 20, 30, 40]
print(a[0:3])
print(a[2:])
print(a[:4])
print(a[:-1])
```
