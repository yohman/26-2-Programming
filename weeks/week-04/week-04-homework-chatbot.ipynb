## まず動かす
次のCellは、2種類のメッセージに返す小さなBotです。

1. 実行して「こんにちは」「つかれた」「宇宙」と入力。
2. 「終了」で終える。
3. テーマと言葉を自分のものに変え、**三つ以上の返答パターン**に育てよう。

APIキー・外部ファイル・追加インストールは不要。

```python
print("放課後Botです。話しかけてね。終える時は「終了」と入力。")

while True:
    message = input("あなた：").strip()

    if message == "終了":
        print("Bot：またね！")
        break
    elif message == "":
        print("Bot：何か一言、入力してね。")
    elif "こんにちは" in message:
        print("Bot：こんにちは！今日はどんな一日？")
    elif "つかれた" in message or "疲れた" in message:
        print("Bot：ちょっと休憩しよう。好きな曲を一曲聴いてみる？")
    else:
        print("Bot：まだその言葉は分からないな。別の言い方で話してみて。")
```

### 読み方
- `input()`で受け取り、`.strip()`で前後の空白を除く。
- `in`で文章に言葉が含まれるか調べる。
- `elif`は上から順。`else`が想定外の入力を受け止める。
- `while True:`は会話を繰り返す準備済みの部分。`break`で終わる。繰り返しは次週詳しく学ぶので、今回はそのまま使ってOK。

これは自分で書いたルールで返すBotです。チャット型AIのAPIは使いません。
