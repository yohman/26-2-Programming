## 1｜Botのプロフィール
**辞書**に名前・役割・顔を、**リスト**に好きなものを三つ入れる。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
bot = {
    "name": "コト",
    "face": "🤖",
    "role": "放課後の話し相手",
    "likes": [
        "音楽",
        "散歩",
        "おにぎり",
    ],
}

faces = {
    "normal": "🤖",
    "happy": "🤖✨",
    "tired": "🤖💤",
}

print(f"{bot['face']} {bot['name']}｜{bot['role']}")
```

## 2｜三つの気分、いろいろな返事
**normal（ふつう）・happy（元気）・tired（おつかれ）**の三つだけ。
気分ごとのリストに、返事を八つずつ用意しています。

- `responses["happy"][0]`：happyの最初の返事。
- `random.choice(responses["happy"])`：happyのリストから一つランダムに選ぶ。

`random` はPythonに付いている道具。インストールは不要です。同じ返事が続くこともあります。

```python
import random

responses = {
    "normal": [
        f"こんにちは！{bot['name']}です。",
        f"{bot['role']}に、何か話してみて。",
        "今日はどんな一日だった？",
        "その話、もう少し聞かせて。",
        f"私は{bot['likes'][0]}が好き。あなたは？",
        f"今日は{bot['likes'][2]}の気分！",
        "知らない話題もあるけど、聞いてみたいな。",
        "最近、小さな発見はあった？",
    ],
    "happy": [
        "その元気、少し分けてほしい！",
        "いいね！何があったの？",
        "今日は心の充電が100％だね。",
        "その調子で、小さな挑戦を一つ！",
        "うれしいことは、誰かに話したくなるね。",
        "お祝いに、好きな曲を一曲どう？",
        f"私なら{bot['likes'][1]}に出かけたい！",
        "その気分、明日の自分にも届けよう。",
    ],
    "tired": [
        "今日は省エネモードでいこう。",
        "まず水を一口。急がなくて大丈夫。",
        "おつかれさま。少し休憩しない？",
        "全部じゃなくて、一つできたら十分。",
        "好きな曲一曲ぶん、ひと休み。",
        "肩の力を抜いて、深呼吸。",
        f"{bot['likes'][2]}でも食べて休もう。",
        "私も充電が必要。いっしょに休もう。",
    ],
}

print(responses["happy"][0])
print(random.choice(responses["happy"]))
```

## 3｜気分を選んで、返す
「こんにちは」「元気じゃない」「元気」「楽しい」「つかれた」「宇宙」を試そう。
**「終了」でCellの実行を終える。**

条件で `mood` を決める → `responses[mood]` からランダムに一つ選ぶ → 表示する。

自分で関数を定義する必要はありません。
`while True` と `break` は会話を続けるための準備済み部分。今週はそのまま使ってOK。

```python
print(f"{bot['face']} {bot['name']}：話しかけてね。『終了』でおしまい。")

while True:
    message = input("あなた：").strip()

    if message == "終了":
        print(f"{bot['face']} {bot['name']}：またね！")
        break

    if "元気じゃない" in message or "つかれ" in message or "疲れ" in message:
        mood = "tired"
    elif "元気" in message or "楽しい" in message:
        mood = "happy"
    else:
        mood = "normal"

    reply = random.choice(responses[mood])
    print(f"あなた：{message}")
    print(f"{faces[mood]} {bot['name']}：{reply}")
    print()
```

> **Aha!** 「元気じゃない」にも「元気」が含まれる。だから、具体的な条件を先に書く。二つの条件の順番を入れ替えると、返事はどう変わる？
