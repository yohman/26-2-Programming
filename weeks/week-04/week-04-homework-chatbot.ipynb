```python
bot = {
    "name": "コト",
    "role": "放課後の話し相手",
    "likes": ["音楽", "散歩", "おにぎり"],
    "color": "#285c73",
}

print(f"{bot['name']}｜{bot['role']}")
print(f"好きなもの：{bot['likes']}")
```

## 2｜返事の引き出し
**話題 → 返事のリスト**。同じ話題に二つずつ返事を用意しています。

`responses["greeting"][0]` は「greetingの引き出しにある、最初の返事」。
**三つ以上の話題の返事**を、自分のBotらしく書き換えよう。

```python
responses = {
    "greeting": ["やっほー！今日はどんな一日？", "来てくれてうれしい。何か話そう！"],
    "tired": ["今日は省エネモードでいこう。まず水を一口。", "おつかれさま。好きな曲一曲ぶん、休憩しない？"],
    "happy": ["その元気、少し分けてほしい！", "いいね！今日よかったことを一つ教えて。"],
    "name": [f"{bot['name']}です。{bot['role']}を担当しています！", f"名前は{bot['name']}。覚えてくれた？"],
    "likes": [f"{bot['likes'][0]}と{bot['likes'][1]}が好き！あなたは？", f"今日は{bot['likes'][2]}の気分。"],
    "joke": ["ロボットが休む日は？……充電日。", "好きなパンは？……キーボードのスペースパン。ちょっと無理があるね。"],
    "help": ["こんにちは・元気・つかれた・名前・好き・ジョーク、と話しかけてね。終了でおしまい。", "得意なのは気分の話と自己紹介。未知の話題は、これから勉強！"],
    "empty": ["無言の会話もいいけど、一言もらえるとうれしい。", "聞こえなかったな。何か入力してみて。"],
    "unknown": ["その話はまだ勉強中。『ヘルプ』で得意な話題を見てね。", "分かったふりはしないよ。別の言い方で教えて！"],
    "bye": ["またね！続きは次の充電のあとで。", "お話ししてくれてありがとう。また遊びにきてね！"],
}

print(responses["greeting"][0])
```

## 3｜Botの顔（表示用・完成済み）
**このCellはそのまま実行でOK。** Notebookの中にロボットと吹き出しを表示します。

`show_bot(...)` は表示係。関数・HTML・SVGは今回の必須範囲ではありません。自分で書くのは、プロフィール・返事・次の条件分岐です。

```python
from IPython.display import HTML, display
from html import escape

motion = ""  # 発展Challengeでアニメーションを追加できる場所

def show_bot(profile, message, reply, mood):
    mouths = {
        "normal": "M32 61 H56",
        "happy": "M31 57 Q44 73 57 57",
        "sleepy": "M35 64 H53",
    }
    color = escape(profile["color"], quote=True)
    face = f"""<svg class="bot-face" viewBox="0 0 88 88" width="88" height="88" role="img" aria-label="ロボットの顔">
      <path d="M44 10 V21" stroke="{color}" stroke-width="4"/><circle cx="44" cy="8" r="5" fill="#d6ad47"/>
      <rect x="16" y="22" width="56" height="52" rx="9" fill="#eef4f6" stroke="{color}" stroke-width="3"/>
      <path d="M10 40 V55 M78 40 V55" stroke="{color}" stroke-width="6"/>
      <circle cx="32" cy="43" r="5" fill="{color}"/><circle cx="56" cy="43" r="5" fill="{color}"/>
      <path d="{mouths[mood]}" fill="none" stroke="{color}" stroke-width="3" stroke-linecap="round"/>
    </svg>"""
    display(HTML(f"""{motion}<div class="bot-card" style="display:flex;gap:18px;align-items:flex-start;padding:20px;background:#f5f8fa;color:#24333e;font-family:sans-serif;max-width:620px">
      <div style="flex:none">{face}</div><div>
      <strong>{escape(profile["name"])}</strong><small style="display:block;color:#546875">{escape(profile["role"])}</small>
      <p style="color:#546875">あなた：{escape(message)}</p>
      <p style="margin:0;line-height:1.8;font-size:17px;white-space:pre-wrap">{escape(reply)}</p>
      </div></div>"""))

show_bot(bot, "こんにちは", responses["greeting"][0], "happy")
```

## 4｜会話してみよう
「こんにちは」「元気じゃない」「元気」「好き」「ジョーク」「宇宙」「ヘルプ」を試そう。**「終了」でCellの実行を終える。**

`topic` で引き出しを決め、`responses[topic][0]` で返事を取り出します。
`while True` と `break` は会話を続けるための準備済み部分。今週はそのまま使ってOK。

```python
print(f"{bot['name']}：話しかけてね。『ヘルプ』で話題、『終了』でおしまい。")

while True:
    message = input("あなた：").strip()
    mood = "normal"

    if message in ["終了", "ばいばい", "bye"]:
        topic = "bye"
    elif message == "":
        topic = "empty"
    elif "元気じゃない" in message or "つかれた" in message or "疲れた" in message:
        topic = "tired"
        mood = "sleepy"
    elif "元気" in message:
        topic = "happy"
        mood = "happy"
    elif "こんにちは" in message or "やっほー" in message:
        topic = "greeting"
        mood = "happy"
    elif "名前" in message:
        topic = "name"
    elif "好き" in message:
        topic = "likes"
    elif "ジョーク" in message:
        topic = "joke"
        mood = "happy"
    elif "ヘルプ" in message:
        topic = "help"
    else:
        topic = "unknown"

    reply = responses[topic][0]
    show_bot(bot, message, reply, mood)

    if topic == "bye":
        break
```

> **Aha!** 「元気じゃない」にも「元気」が含まれる。だから、具体的な条件を先に書く。二つの条件の順番を入れ替えると、返事はどう変わる？
