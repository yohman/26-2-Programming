## 1｜Botのプロフィール
**辞書**に名前・役割・顔を、**リスト**に好きなものを三つ入れる。まず自分のキャラクターに変えよう。

Pythonの `[]` は **list（リスト）**。今回はNumPyのarrayは使いません。

```python
bot = {
    "name": "コト",
    "face": "🤖",
    "role": "放課後の話し相手",
    "likes": ["音楽", "散歩", "おにぎり"],
}

faces = {
    "normal": "🤖",
    "happy": "🤖✨",
    "sleepy": "🤖💤",
}

print(f"{bot['face']} {bot['name']}｜{bot['role']}")
print(f"好きなもの：{bot['likes']}")
```

## 2｜返事の引き出し
**話題 → 返事のリスト**。同じ話題に二つずつ返事を用意しています。

`responses["greeting"][0]` は「greetingの引き出しにある、最初の返事」。
**三つ以上の話題の返事**を、自分のBotらしく書き換えよう。

```python
responses = {
    "greeting": ["やっほー！今日はどんな一日？", "来てくれてうれしい。何か話そう！"],
    "tired": ["今日は省エネモードでいこう。まず水を一口。", "おつかれさま。好きな曲一曲ぶん、休憩しない？"],
    "happy": ["その元気、少し分けてほしい！", "いいね！今日よかったことを一つ教えて。"],
    "name": [f"{bot['name']}です。{bot['role']}を担当しています！", f"名前は{bot['name']}。覚えてくれた？"],
    "likes": [f"{bot['likes'][0]}と{bot['likes'][1]}が好き！あなたは？", f"今日は{bot['likes'][2]}の気分。"],
    "joke": ["ロボットが休む日は？……充電日。", "好きなパンは？……キーボードのスペースパン。ちょっと無理があるね。"],
    "help": ["こんにちは・元気・つかれた・名前・好き・ジョーク、と話しかけてね。終了でおしまい。", "得意なのは気分の話と自己紹介。未知の話題は、これから勉強！"],
    "empty": ["無言の会話もいいけど、一言もらえるとうれしい。", "聞こえなかったな。何か入力してみて。"],
    "unknown": ["その話はまだ勉強中。『ヘルプ』で得意な話題を見てね。", "分かったふりはしないよ。別の言い方で教えて！"],
    "bye": ["またね！続きは次の充電のあとで。", "お話ししてくれてありがとう。また遊びにきてね！"],
}

print(responses["greeting"][0])
```

## 3｜会話してみよう
「こんにちは」「元気じゃない」「元気」「好き」「ジョーク」「宇宙」「ヘルプ」を試そう。**「終了」でCellの実行を終える。**

- 条件分岐で `topic` を決める。
- `responses[topic][0]` で返事を取り出す。
- `faces[mood]` で顔を選び、`print()` で表示する。

自分で関数を定義する必要はありません。
`while True` と `break` は会話を続けるための準備済み部分。今週はそのまま使ってOK。

```python
print(f"{bot['face']} {bot['name']}：話しかけてね。『ヘルプ』で話題、『終了』でおしまい。")

while True:
    message = input("あなた：").strip()
    mood = "normal"

    if message in ["終了", "ばいばい", "bye"]:
        topic = "bye"
    elif message == "":
        topic = "empty"
    elif "元気じゃない" in message or "つかれた" in message or "疲れた" in message:
        topic = "tired"
        mood = "sleepy"
    elif "元気" in message:
        topic = "happy"
        mood = "happy"
    elif "こんにちは" in message or "やっほー" in message:
        topic = "greeting"
        mood = "happy"
    elif "名前" in message:
        topic = "name"
    elif "好き" in message:
        topic = "likes"
    elif "ジョーク" in message:
        topic = "joke"
        mood = "happy"
    elif "ヘルプ" in message:
        topic = "help"
    else:
        topic = "unknown"

    reply = responses[topic][0]
    print(f"あなた：{message}")
    print(f"{faces[mood]} {bot['name']}：{reply}")
    print()

    if topic == "bye":
        break
```

> **Aha!** 「元気じゃない」にも「元気」が含まれる。だから、具体的な条件を先に書く。二つの条件の順番を入れ替えると、返事はどう変わる？
