### 同じ処理を手で繰り返す / Repeating by hand

`count += 1`は`count = count + 1`と同じ。`'{} sheep'.format(count)`は、`{}`に`count`の値を入れる。`f'{count} sheep'`とも書ける。

````python
# これもfor loopを使わないけど変数を使う
count = 0
count += 1
print('{} sheep'.format(count))
count += 1
print('{} sheep'.format(count))
count += 1
print('{} sheep'.format(count))
count += 1
print('{} sheep'.format(count))
print('Good night')
````

　上記のように、ある1部分だけ数値を変更した命令文、今回の場合は
```
count += 1
print('{} sheep'.format(count))
```
を繰り返し実行したいという場面は実際に多いです。実際のデータ分析においては、サッカーデータの分析などでは

1.   年ごとにJリーグ選手の平均年俸を計算する
2.   チームごとに選手の平均年俸を計算する

のように、**対象だけ変えて同じ分析(処理)を行う**というようなケースも多いです。  
　そのような時に、年の数だけ分析を実行したいプログラムを書いてもいいですが、正直面倒です。手作業でコードを書くと、ミスも起きたりします。そこで、**繰り返し**という処理があります。

上のプログラムをfor文を使って書いてみると以下のようなプログラムになります。

````python
# リストを作る
numbers = [1,2,3,4]

# loopでアウトプットを出す
for count in numbers:
    print('{} sheep'.format(count))
print('Good night')
````

　for文の基本的な使い方は、上記のような  
  ```
　　for 変数 in リスト:  
　　　　　繰り返し処理1  
　　　　　繰り返し処理2  
　　　・・・
  ```
という使い方です。リストだけでなく、タプルやディクショナリ、セットなどでも大丈夫です。また、for文直後のブロックを繰り返し実行しますが、このブロックを**forブロック**といいます。if文の時と同様に、**forブロックはインデントによって、ブロックを指定します**。

　上記の例を見るとわかるように、for文の動作イメージは、  
　　1.リストの1つ目の値が変数に代入されて、繰り返し処理を順に実行していく  
　　2.1の繰り返し処理が最後まで終わったら、リストの2つ目の値が変数に代入されて、繰り返し処理を順に実行していく  
　　3.この流れをリストの最後の値まで繰り返す  
という感じです。  
```
for 変数 in リスト:
```
の`変数`の名前は何でもよいです。わかりやすいものにしましょう。

````python
# example with text

# create a list of names
coders = ['yoh','miyu','ryo','airi','shun']

# loop and print out each name
for coder in coders:
    print(coder + " loves coding")
````

　リストの要素が4つ程度であれば上記のような書き方で問題ないのですが、1～100までになると、リスト（ここではnumbers）を用意するのも大変です。そこで、range関数を使います。range関数を使って、`range(n)`とすると0以上n**未満**の数列（リストみたいなもの）が生成されます。
例えば、`range(3)`とすると 
`0, 1, 2`
という数列が生成されます。
`range`の引数が1つだけだと0から始まる数列が生成されますが、
```
range(start, end)
```
のように2つの引数を与えると、start以上end**未満**までの数列が生成されます。
例えば、`range(1,5)`とすると数列`1, 2, 3, 4`が生成されます。

この場合、
```
for 変数 in range(n):
  繰り返し処理1  
  繰り返し処理2  
  ・・・
```
や
```
for 変数 in range(start,end): 
  繰り返し処理1  
  繰り返し処理2  
  ・・・
```
のようにすればOK!

````python
# range example
for i in range(10):
    print(i)
````

### 例2 / Example 2: inputとrange

### 例4 / Example 4: inputで合計を変える

## 決まった回数の繰り返し / Repeating a fixed number of times

ここまで説明してきたように、for文は**決まった回数の繰り返し**でよく使われます。つぎのプログラムの中で変数 `i` は繰り返し処理に使われていないことに注意しましょう。

### for文によるリストの作成
　最後にforを使ったリストの作成方法について紹介します。ここまでは、繰り返し処理の中で変数の値を変えて出力をしたり、sum関数を使用せずに、リストの要素を合計するなどを例題としてやりました。実はfor文を使うことで、リストの作成も行えます。

　例えば、`[1, 2, 3, 4, 5]`というリストがあるとして、このリストの各要素を2乗したリストを作成してみます。

````python
# 元のリスト / Original list
values = [1, 2, 3, 4, 5]

# 空のリストから始める / Start with an empty list
squares = []
for value in values:
    squares.append(value ** 2)

print(squares)
````

## まとめ
同じ処理を実行するときに、命令文を何度も書くのは煩雑なので、繰り返し文を用います。繰り返し文の1つである`for`の使い方は
```
for 変数 in リスト:
  繰り返し処理
```
であり、
```
for 変数 in range(start,end):
　　　繰り返し処理
```
という書き方もよく使われます。`for 変数 in リスト`や`for 変数　in range(start,end)`という書き方からもわかるように、**「for文は繰り返し回数が明らかな時に使う」「繰り返し回数はリストの要素の数」**と覚えておけば、だいたいOK!
