## 2. 絞り込みと累積
条件を満たす値だけを新しいリストに入れます。`append()`が`if`の中にある理由を説明します。

```python
values = [3, 8, 2, 9, 5]
selected = []
for value in values:
    if value >= 5:
        selected.append(value)
print(selected)
```
