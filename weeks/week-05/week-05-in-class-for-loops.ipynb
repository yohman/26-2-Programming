## 1. リストから一つずつ取り出す

`for name in names:` は、`names` の一件ずつを `name` に入れて、字下げした処理を繰り返します。

## 2. `range()` で決まった回数を繰り返す

`range(start, end)` は **startからendの手前まで** の数を作ります。`range(1, 5)` の最後の数は4です。

実行前に予想：`range(5)` はどの数から始まり、どの数で終わりますか？

```python
for number in range(1, 5):
    print(f'Round {number}')

# H: range(5) に変えて、出力をMarkdown Cellに説明する。
```

## 3. 累積変数：一周ごとに合計を更新する

`total` はループの前で0から始めます。各周回で、前の合計に次の値を足します。

実行前に予想：最後の `total` はいくつですか？ `total = 0` をループの中に入れたら、何が変わりますか？

```python
scores = [3, 8, 2, 9]
total = 0

for score in scores:
    total += score
    print(f'after {score}: {total}')

print(f'Final total: {total}')
```

## 4. 絞り込み：条件を通った値だけを残す

空のリスト `passing` を用意し、`if` を通った値だけを `append()` で追加します。`append()` は **ifの中に字下げ** します。

実行後、合格ラインを60から80に変えてもう一度実行します。

```python
scores = [43, 61, 78, 55, 92]
passing = []

for score in scores:
    if score >= 60:
        passing.append(score)

print(passing)
```

## 6. `japan.csv` をデータとして読む

`japan.csv` をこのNotebookと同じフォルダに置きます。最初に行・列・値の型を確認し、使う列を決めます。

このデータでは、各行が一つの都道府県の記録です。リストで練習した「一件ずつ同じ処理」が、そのままデータにも使えます。

```python
import pandas as pd

df = pd.read_csv('japan.csv')
display(df[['都道府県', '総人口', '合計特殊出生率']].head())
print(df.shape)
print(df[['総人口', '合計特殊出生率']].dtypes)
```

## 7. 一件のデータから、一つの説明をつくる

まず三都道府県だけを小さなリストにして、各記録から読みやすいラベルを作ります。

実行前に予想：リストに追加されるラベルは何個ですか？

```python
sample = df.loc[
    df['都道府県'].isin(['北海道', '東京都', '大阪府']),
    ['都道府県', '総人口', '合計特殊出生率']
].to_dict('records')

labels = []
for prefecture in sample:
    label = f"{prefecture['都道府県']}: {prefecture['総人口']:,} people"
    labels.append(label)

print(*labels, sep='\n')
```

## 8. データを絞り込む

今度は全都道府県を一行ずつ調べ、総人口200万人以上の都道府県名だけを集めます。

実行後、しきい値を `6_000_000` に変えます。表示される名前を予想してから実行してください。

```python
large_prefectures = []

for _, row in df.iterrows():
    if row['総人口'] >= 2_000_000:
        large_prefectures.append(row['都道府県'])

print(large_prefectures)
print('count:', len(large_prefectures))
```

## 9. 全行を使って合計をつくる

`df['総人口']` は、総人口だけを順番に取り出せる列です。各人口を `total_population` に加えます。

最後に、Pandasの `sum()` とも比べます。同じ結果か確認してください。

```python
total_population = 0

for population in df['総人口']:
    total_population += population

print(f'Loop total: {total_population:,}')
print(f'pandas sum: {df["総人口"].sum():,}')
```
