## メニューは、辞書のリスト
nameは商品名、priceは値段（円）。まず4品から。

name is the item name; price is the cost in yen. Start with four items.

```python
menu = [
    {"name": "Onigiri", "price": 150},
    {"name": "Sandwich", "price": 320},
    {"name": "Curry", "price": 550},
    {"name": "Udon", "price": 450}
]
```

## forで、1品ずつ表示
itemには、メニューの商品が一つずつ入る。

item receives one menu entry at a time.

```python
for item in menu:
    print(item["name"], item["price"])
```

## ifで、予算と比べる
priceがbudget以下なら表示する。500円で買えるのはどれ？

Print an item if its price is at or below budget. Which items can you buy with ¥500?

```python
budget = 500
for item in menu:
    if item["price"] <= budget:
        print(item["name"], item["price"])
```

## 2品を選ぶ
menu[0]は最初、menu[1]は2番目。orderに、選んだ商品を入れる。

menu[0] is the first item; menu[1] is the second. Put your selected items in order.

```python
order = [menu[0], menu[1]]
for item in order:
    print(item["name"], item["price"])
```

## 合計と残金を計算
totalは0から始め、選んだ商品の値段を足す。残金は予算−合計。

Start total at zero and add each selected price. Money left is budget minus total.

```python
budget = 1000
total = 0
for item in order:
    total += item["price"]
print("Total:", total, "yen")
print("Money left:", budget - total, "yen")
```

## メニューをCSVから読む
csv.DictReaderは各行を辞書として読む。priceは文字列なのでintで数値にする。CSVとNotebookは同じフォルダへ。

csv.DictReader reads each row as a dictionary. Convert price from text to a number with int. Keep the CSV beside your notebook.

```python
import csv

with open("student-menu.csv") as file:
    menu = list(csv.DictReader(file))
for item in menu:
    item["price"] = int(item["price"])
print(len(menu), "items")
```

## 予算が足りなかったら？
