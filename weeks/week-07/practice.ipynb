## for と while
forはリストの各要素を調べる。whileは条件がTrueの間続く。どちらも状態の変化を追う。

for visits items; while continues while a condition is true. Trace the changing state.

```python
coins = [100, 30, 50]
paid = 0
for coin in coins:
    if coin in [10, 50, 100]:
        paid += coin
print(paid)
```

## まず、4つだけ
1件は辞書、全体はリスト。価格と時間は練習用の架空の値。実際の店の価格ではありません。

Each item is a dictionary, the menu a list. Prices and times are fictional practice data.

```python
menu = [
    {"name": "Onigiri", "price": 150, "minutes": 3},
    {"name": "Sandwich", "price": 320, "minutes": 5},
    {"name": "Curry", "price": 550, "minutes": 12},
    {"name": "Udon", "price": 450, "minutes": 10}
]
```

## 予算内の候補
<=は「以下」。150円の予算なら、150円の商品も含める。

<= includes an item that costs exactly the budget.

```python
budget = 500
for item in menu:
    if item["price"] <= budget:
        print(item["name"], item["price"])
```

## 候補と、買い物かごは違う
候補を全部買うわけではない。選んだ2品の合計をループで足す。

You do not buy every eligible option. Add the prices of your selected items.

```python
basket = [menu[0], menu[1]]
total = 0
for item in basket:
    total += item["price"]
print("Total:", total)
print("Remaining:", 1000 - total)
```
