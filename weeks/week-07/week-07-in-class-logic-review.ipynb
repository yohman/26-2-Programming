## forで、コインを順に足す
今度はforでリストを順に読む。paid += coinは金額を足して更新する。前のwhileは、150円に届いた時点で止まった。

A for loop visits each coin in the list. paid += coin adds to the running total. The previous while stopped once payment reached ¥150.

```python
coins = [100, 50]
paid = 0
for coin in coins:
    paid += coin
    print("Paid:", paid)
```

## 本題：メニューから選ぶ
nameは名前、priceは円、minutesは使う時間。1件は辞書、全体はリスト。価格・時間は架空の練習用データ。

Now plan your time after class. name identifies an item, price is yen and minutes is time spent. Each item is a dictionary in a list; values are fictional.

```python
menu = [
    {"name": "Onigiri", "price": 150, "minutes": 3},
    {"name": "Sandwich", "price": 320, "minutes": 5},
    {"name": "Curry", "price": 550, "minutes": 12},
    {"name": "Udon", "price": 450, "minutes": 10}
]
```

## 予算内の候補
<=は「以下」。150円の予算なら、150円の商品も含める。

<= includes an item that costs exactly the budget.

```python
budget = 500
for item in menu:
    if item["price"] <= budget:
        print(item["name"], item["price"])
```

## 選んだプランの合計
例はおにぎり＋サンドイッチ。金額と時間をforで足し、予算と時間の両方を確認。表示された候補を全部選ぶ必要はない。

This example selects onigiri and a sandwich. Add their cost and time with for, then check both limits. You choose which options go into your plan.

```python
budget = 1000
max_minutes = 45
plan = [menu[0], menu[1]]
total_price = 0
total_minutes = 0
for item in plan:
    total_price += item["price"]
    total_minutes += item["minutes"]
print("Cost:", total_price, "yen")
print("Money left:", budget - total_price, "yen")
print("Time:", total_minutes, "minutes")
print("Within budget:", total_price <= budget)
print("Within time:", total_minutes <= max_minutes)
```
