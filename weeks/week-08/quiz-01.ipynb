## 小テスト①問題1-3（5点）

キーボードからいくつかの文字を半角スペース区切りで入力する。そのとき、最後から1つ前の要素を出力するプログラムを作成せよ。
例えば、
```
a b c d e
```
が入力されたとき、`d` を出力する。キーボードからは2つ以上の文字が入力されることを前提としてよい。

プログラムの最初に
```
mylist = input().split()
```
と書くと、キーボードから半角スペース区切りで入力されたいくつかの文字がリスト型変数 `mylist` に入力される。

Hint: リストの最後から1つ前の要素は `mylist[-2]` で参照できる。

## 小テスト①問題1-4（5点）

キーボードから4つの文字を半角スペース区切りで入力する。入力された文字を逆の順番で出力するプログラムを作成せよ。

例えば、キーボードから
```
1 2 3 4
```
が入力されたとき
```
4 3 2 1
```
と出力する。

入力される整数の個数が4であることは前提としてよい。

プログラムの先頭に
```
mylist = input().split()
```
と書くと、キーボードから半角スペース区切りで入力された文字が、リスト型変数 `mylist` に代入される。

Hint: リストの要素を逆順に参照するにはスライスを用いる。たとえば、`mylist[::-1]` は `mylist` の要素を逆順にしたリストを表す。
