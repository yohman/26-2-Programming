## 同じ引き算を一つに
defは処理の定義。呼ぶまで動かない。budgetとpriceは受け取る値。returnは結果を呼び出し元へ返す。

def defines the operation; calling runs it. Parameters receive values; return sends the result back.

```python
def remaining(budget, price):
    balance = budget - price
    return balance

print(remaining(1000, 470))
print(remaining(700, 470))
```

## 候補0件を伝える
空のリストはFalseとして扱える。not chosenなら一件もなかった。

An empty list is falsey; not chosen means no item matched.

```python
chosen = []
for item in menu:
    if item["price"] <= 0:
        chosen.append(item["name"])
if not chosen:
    print("No options. Try another budget.")
```

## 結果を、目で照合する
assertは条件が違えば止まる確認。まず手で計算してから書く。

assert stops when a check is false. Calculate the expected answer first.

```python
assert remaining(1000, 470) == 530
assert remaining(470, 470) == 0
print("Two checks passed")
```

## 集計も、同じfor
この3行は説明用の小さな例。完全な事件データは選択Notebookで読み込む。

These three rows are a teaching example. The alternative notebook loads the actual incident CSV.

```python
places = ["Parking", "Road", "Parking"]
counts = {}
for place in places:
    counts[place] = counts.get(place, 0) + 1
print(counts)
```
