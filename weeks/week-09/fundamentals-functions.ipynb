```python
menu = [
    {"name": "Onigiri", "price": 150, "minutes": 3},
    {"name": "Sandwich", "price": 320, "minutes": 5},
    {"name": "Curry", "price": 550, "minutes": 12},
    {"name": "Udon", "price": 450, "minutes": 10}
]
```

## まず、知っている処理
価格を調べ、候補を集める。この処理を名前で呼べるようにしたい。

Inspect prices and collect matching names. Give this process a name.

```python
chosen = []
for item in menu:
    if item["price"] <= 500:
        chosen.append(item["name"])
print(chosen)
```

## defは、まだ実行しない
字下げ部分を関数として保存する。名前の後の()で初めて呼び出す。

The indented body is defined here. Parentheses on a call run it.

```python
def remaining(budget, price):
    balance = budget - price
    return balance

print("Function defined; body not called yet")
```

## 引数を渡す
1000→budget、470→price。引数は呼び出す側の値、仮引数は関数内の受け口。

1000 becomes budget; 470 becomes price. Arguments are supplied values; parameters are the names receiving them.

```python
balance = remaining(1000, 470)
print(balance)
```

## printとreturnは役割が違う
printは画面へ表示。returnは次の計算で使う値を返す。returnがなければNoneが返る。

print displays; return supplies a reusable result. With no explicit return, the result is None.

```python
def show_balance(budget, price):
    print(budget - price)

result = show_balance(1000, 470)
print("result:", result)
```

## 関数の中にfor
関数全体で一つの予算を処理。chosen=[]は毎回の呼び出しで作り直す。

One call processes one budget. chosen=[] creates a fresh list each time.

```python
def affordable(items, budget):
    chosen = []
    for item in items:
        if item["price"] <= budget:
            chosen.append(item["name"])
    return chosen

print(affordable(menu, 500))
```

## 外側のforから関数を呼ぶ
forが予算を一つずつ取り出す。関数がその予算の候補を返す。

The outer for supplies one budget; the function returns that budget’s choices.

```python
for budget in [500, 700, 1000]:
    names = affordable(menu, budget)
    print(budget, names)
```

## returnが早すぎる
この例では最初の品だけでreturnする。修正前の結果を予想する。

This version returns after the first item. Predict the bug before fixing it.

```python
def too_early(items, budget):
    chosen = []
    for item in items:
        if item["price"] <= budget:
            chosen.append(item["name"])
        return chosen

print(too_early(menu, 500))
```

## 空の入力も、結果を返す
returnをループの外へ置くと、空のリストでも[]を返す。

With return after the loop, an empty input still returns [].

```python
assert affordable([], 500) == []
assert affordable(menu, 149) == []
assert affordable(menu, 150) == ["Onigiri"]
print("Three checks passed")
```
