## 見本の名前 / Sample names

```python
names = ["Haru", "Yui", "Ren"]

print(f"HELLO, {names[0]}!")
print(f"HELLO, {names[1]}!")
```

## 関数を定義 / Define a function

defは処理に名前を付ける。nameは、呼び出し時に値を受け取る仮引数。字下げした行は呼ぶまで動かない。

def names the operation. name is the parameter that receives a value. The indented body waits until a call.

```python
def name_tag(name):
    return f"HELLO, {name}!"
```

## 呼び出しと戻り値 / Call and return

```python
tag = name_tag("Haru")
print(tag)
```

## 表示と戻り値 / Display and return

```python
def show_tag(name):
    print(name_tag(name))

result = show_tag("Yui")
print(result)
```

## forの中から呼ぶ / Call inside a loop

```python
for student in names:
    print(name_tag(student))
```

## 都市ラベル / Weather label

```python
def weather_label(city, icon, high, low):
    return f"{icon} {city}\n↑{high}°C  ↓{low}°C"

print(weather_label("柏", "☁️", 16.6, 6.1))
```

## 都市リストと気象データ / City list and weather data

```python
forecast = {'柏': ('☁️', 16.6, 6.1),
 '札幌': ('☁️', 10.8, -0.1),
 '大阪': ('☁️', 18.2, 7.2),
 '那覇': ('🌦️', 25.9, 21.1)}

cities = ["柏", "札幌", "大阪", "那覇"]

print(forecast[cities[0]])
```

## 都市ごとに関数を呼ぶ / Call once per city

```python
for city in cities:
    icon, high, low = forecast[city]
    print(weather_label(city, icon, high, low))
```

## バッジを組み立てる / Build a badge

```python
from html import escape

def weather_badge(city, icon, high, low):
    label = weather_label(city, icon, high, low)
    safe_label = escape(label)
    style = ("padding:.75rem;background:#e8eff3;"
             "border-top:3px solid #315876;white-space:pre-line")
    return f'<div style="{style}">{safe_label}</div>'
```

```python
from IPython.display import HTML, display
display(HTML(weather_badge("柏", "☁️", 16.6, 6.1)))
```

## 都市をまとめて表示 / Compose a city panel

```python
def weather_panel(cities, forecast):
    badges = []
    for city in cities:
        icon, high, low = forecast[city]
        badges.append(weather_badge(city, icon, high, low))

    style = "display:grid;grid-template-columns:repeat(2,1fr);gap:8px"
    return f'<div style="{style}">{"".join(badges)}</div>'
```

```python
from IPython.display import HTML, display
display(HTML(weather_panel(cities, forecast)))
```
