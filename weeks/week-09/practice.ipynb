## forの中から呼ぶ / Call inside a loop

```python
# 名前のリストを用意する
names = ["Haru", "Yui", "Ren"]

# 関数を定義する
def greet_name(name):
    return f"Hello, {name}!"

# 1人ずつ関数を呼び、結果をリストに入れる
messages = []
for student in names:
    message = greet_name(student)
    messages.append(message)
messages
```

## マグニチュードで絞り込む / Filter by magnitude

```python
# USGSの39件。各組は (マグニチュード, 深さ km)
quakes = [
    (3.1, 8.19), (2.75310623645782, 16.99), (2.84, 19.21), (4.9, 35), (4.7, 10),
    (3.4, 49.21), (3.2, 65.6), (4.9, 44.91), (4.8, 10), (4.7, 10),
    (4.6, 10), (5, 10), (5.1, 21.95), (2.6, 8.98), (2.83, 21.72),
    (4.4, 124.33), (3.6, 10), (4.8, 162.21), (2.72, 34.29), (6, 20.84),
    (4.4, 10), (2.57, -0.32), (5.1, 28.4), (3.87, 117), (4.8, 65.7),
    (5.3, 18.52), (2.5, 55.5), (2.71, 1.72), (4.5, 10), (4.1, 32.2),
    (6.8, 10), (4.3, 188.82), (4.9, 10), (3.37, 21.94), (2.72, 16.8),
    (2.66, 9.43), (2.47, 18.28), (4.5, 35), (4.7, 10),
]

# 関数を定義する
def strong_quakes(quakes, minimum):
    selected = []
    for quake in quakes:
        magnitude, depth = quake
        if magnitude >= minimum:
            selected.append(quake)
    return selected

# M5以上を選び、件数を見る
selected = strong_quakes(quakes, 5.0)
len(selected)
```

## 記録件数の棒グラフ / Bar chart of reported thefts

```python
# bike.csvで数えた3市の自転車盗記録
counts = {'柏市': 564, '松戸市': 725, '船橋市': 1299}

# 関数を定義する
def bar(count, unit):
    return "█" * (count // unit)

def theft_bar(city, counts):
    count = counts[city]
    return f"{city} ({count:,})\n" + bar(count, 100)

# 1市ずつ関数を呼び、改行を含む結果を表示
for city in counts:
    result = theft_bar(city, counts)
    print(result)
```

## 65歳以上の割合 / Share aged 65+

```python
# japan.csvから4都県の65歳以上人口と総人口
prefectures = {
    "千葉県": {'65+': 1753000, 'total': 6266000},
    "東京都": {'65+': 3202000, 'total': 14038000},
    "秋田県": {'65+': 359000, 'total': 930000},
    "沖縄県": {'65+': 344000, 'total': 1468000},
}

# 関数を定義する
def senior_share(row):
    ratio = row["65+"] / row["total"]
    return round(ratio * 100, 1)

# 各都県の割合を辞書にまとめる
shares = {}
for name, row in prefectures.items():
    shares[name] = senior_share(row)
shares
```

## バッジを組み立てる / Build a badge

```python
# HTML表示に使う escape と柏の天気
from html import escape

city = "柏"
icon, high, low = ('☁️', 16.6, 6.1)

# ラベルを作り、バッジ用のHTMLにする
def weather_label(city, icon, high, low):
    return f"{icon} {city}\n↑{high}°C  ↓{low}°C"

def weather_badge(city, icon, high, low):
    label = weather_label(city, icon, high, low)
    safe_label = escape(label)
    style = ("padding:.75rem;background:#e8eff3;"
             "border-top:3px solid #315876;white-space:pre-line")
    return f'<div style="{style}">{safe_label}</div>'

# 関数を呼び出す
badge = weather_badge(city, icon, high, low)
```

```python
# NotebookでHTMLを表示する
from IPython.display import HTML, display
display(HTML(badge))
```

## 都市をまとめて表示 / Compose a city panel

```python
# HTML表示に使う escape と4都市分の天気
from html import escape

forecast = {'柏': ('☁️', 16.6, 6.1),
 '札幌': ('☁️', 10.8, -0.1),
 '大阪': ('☁️', 18.2, 7.2),
 '那覇': ('🌦️', 25.9, 21.1)}

cities = ["柏", "札幌", "大阪", "那覇"]

# ラベル、バッジ、パネルを作る関数
def weather_label(city, icon, high, low):
    return f"{icon} {city}\n↑{high}°C  ↓{low}°C"

def weather_badge(city, icon, high, low):
    label = weather_label(city, icon, high, low)
    safe_label = escape(label)
    style = ("padding:.75rem;background:#e8eff3;"
             "border-top:3px solid #315876;white-space:pre-line")
    return f'<div style="{style}">{safe_label}</div>'

def weather_panel(cities, forecast):
    badges = []
    for city in cities:
        icon, high, low = forecast[city]
        badge = weather_badge(city, icon, high, low)
        badges.append(badge)

    style = "display:grid;grid-template-columns:repeat(2,1fr);gap:8px"
    return f'<div style="{style}">{"".join(badges)}</div>'

# 関数を呼び出す
panel = weather_panel(cities, forecast)
```

```python
# NotebookでHTMLを表示する
from IPython.display import HTML, display
display(HTML(panel))
```
