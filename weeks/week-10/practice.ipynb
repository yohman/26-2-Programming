## 自分の関数を呼ぶ
定義が先、呼び出しが後。Notebookの実行順に頼らず、ファイルを最初から実行する。

Define before calling; the script starts fresh rather than relying on notebook state.

````python
def remaining(budget, price):
    balance = budget - price
    return balance

print(remaining(1000, 470))
````
