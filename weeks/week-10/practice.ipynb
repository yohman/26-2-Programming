````python
# 数字は、ビルの高さ
heights = [2, 5, 3, 6, 4, 2]


def skyline_row(level):
    row = ""
    for height in heights:
        if height >= level:
            row += "[#] "
        else:
            row += "    "
    return row.rstrip()


# 高いところから1段ずつ描く
for level in range(max(heights), 0, -1):
    print(skyline_row(level))
print("=" * (4 * len(heights)))
````

### tiny.py

```python
from time import sleep

spinner = ["|", "/", "-", "\\"]
faces = ["🙂", "😮"]


def tiny_frame(step):
    # % は余り。0〜3、0〜1を繰り返す
    spin = spinner[step % 4]
    face = faces[step % 2]
    dot = " " * step + "."
    return f"SPIN   {spin}\nDOT    {dot}\nEMOJI  {face}"


for step in range(12):
    print("\033[2J\033[H" + tiny_frame(step), flush=True)
    sleep(0.2)
```

    return " " * position + "[==][==][__]>o"

        print("\r" + train_frame(position), end="", flush=True)

    print()


animate_train(18, 0.12)

### route.py

```python
from time import sleep


# 停留所のリストと、今いる停留所の番号を受け取る
def route_frame(stops, current):
    rows = ["CAMPUS SHUTTLE"]
    for i in range(len(stops)):
        marker = "  "
        if i == current:
            marker = "> "
        rows.append(marker + stops[i])
    return "\n".join(rows)


stops = ["Campus", "Library", "Station"]

# 描き直すたびに、次の停留所へ
for current in range(len(stops)):
    print("\033[2J\033[H" + route_frame(stops, current), flush=True)
    sleep(0.65)
```

### signal.py

```python
from time import sleep

lights = ["STOP", "GO", "WAIT"]


def signal_frame(active):
    rows = ["+----------+"]
    for light in lights:
        # 今の状態にだけ星印をつける
        mark = "*" if light == active else "."
        rows.append(f"| ({mark}) {light:<4} |")
    rows.append("+----------+")
    return "\n".join(rows)


for active in ["STOP", "GO", "WAIT", "STOP"]:
    print("\033[2J\033[H" + signal_frame(active), flush=True)
    sleep(0.65)
```

### launch.py

```python
from time import sleep

sky = [
    "  *                    .   ",
    "        .                * ",
    " *                 .       ",
    "          .                ",
    "    .                   *  ",
    "                     .     ",
    "  .                        ",
    "        .               .  ",
    "                           ",
    "                           ",
    "                           ",
    "                           ",
]
rocket = [
    "   ^   ",
    "  / \\  ",
    "  |o|  ",
    " /___\\ ",
    "   *   ",
]
# すべての行を同じ幅にして、空の中央に置く
x = (len(sky[0]) - len(rocket[0])) // 2
launch_row = len(sky) - len(rocket)


def flight_frame(top, message):
    title = message.center(len(sky[0]))
    rows = [title.rstrip()]
    for y in range(len(sky)):
        line = sky[y]
        # ロケットの先端から何行下か
        part = y - top
        if 0 <= part < len(rocket):
            piece = rocket[part]
            end = x + len(piece)
            line = line[:x] + piece + line[end:]
        rows.append(line.rstrip())
    rows.append("-------- LAUNCH PAD -------")
    return "\n".join(rows)


for step in range(10):
    # 上に行くほど、先端の行番号は小さい
    top = max(0, launch_row - step)
    if step == 0:
        message = "READY"
    elif top == 0:
        message = "ORBIT REACHED"
    else:
        message = "LIFTOFF"
    print("\033[2J\033[H" + flight_frame(top, message), flush=True)
    sleep(0.35)
```
