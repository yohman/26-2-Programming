```python
from time import sleep

# 空白の数で、列車の位置を変える
def train_frame(position):
    return " " * position + "[==][==][__]>o"


# 1コマずつ同じ行に描き直す
for position in range(0, 18):
    print("\r" + train_frame(position), end="", flush=True)
    sleep(0.12)

# 最後に改行して、プロンプトと絵を分ける
print()
```

### rain.py

```python
from time import sleep

# 固定の屋根に、雨の行を2行足す
def rain_frame(drops):
    return (
        drops + "\n" + "  " + drops + "\n"
        "        /\\\n"
        "    ___/  \\___\n"
        "      | [] |\n"
        "______|____|______"
    )


# 雨の模様を順番に切り替える
rain = [" .   .    .  .", "  /   /  /   /", "   .   .  .   ."]
for repeat in range(3):
    for drops in rain:
        # 画面を消してから、次のコマを描く
        print("\033[2J\033[H" + rain_frame(drops), flush=True)
        sleep(0.28)
```
