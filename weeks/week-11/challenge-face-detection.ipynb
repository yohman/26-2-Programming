## OpenCVなしの入口 / Cached detections
同じフォルダにdetected-boxes.csvを置く。これは配布画像に対する実際の検出結果。検出自体は再実行していません。

```python
import csv
with open("detected-boxes.csv", encoding="utf-8", newline="") as file:
    boxes = list(csv.DictReader(file))
for box in boxes:
    for key in ["x", "y", "w", "h"]:
        box[key] = int(box[key])
print(len(boxes), "cached boxes")
```
