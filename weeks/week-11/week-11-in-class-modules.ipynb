## 座標は、もう知っているデータ
1つの枠を辞書にすると、メニューと同じリストとして処理できる。ここは説明用の架空座標。

Dictionary boxes can be processed like menu items. These are fictional teaching coordinates.

````python
boxes = [{"x": 40, "y": 30, "w": 80, "h": 90},
         {"x": 180, "y": 50, "w": 60, "h": 70}]
for box in boxes:
    print(box["x"], box["y"], box["w"] * box["h"])
````

## 大きな枠だけ、残す
同じ条件付きループ。min_areaという引数で、選択の基準を外から変える。

The same conditional loop, with min_area as a parameter.

````python
def large_boxes(boxes, min_area):
    kept = []
    for box in boxes:
        if box["w"] * box["h"] >= min_area:
            kept.append(box)
    return kept

print(len(large_boxes(boxes, 5000)))
````
