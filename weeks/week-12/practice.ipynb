## クリック一回で、関数一回
hitは状態を受け取り、次の得点を返す。0秒なら増えない。

hit receives state and returns the next score; no points at zero seconds.

````python
def hit(score, seconds_left):
    if seconds_left > 0:
        return score + 1
    return score

print(hit(0, 3))
print(hit(1, 0))
````

## while Trueで画面を止めない
GUIには入力を待つイベントループがある。root.after(1000, tick)で、1秒後にtickを呼ぶよう予約する。

```python
def tick():
    global seconds_left
    seconds_left -= 1
    if seconds_left > 0:
        root.after(1000, tick)
```

## ループは、モデルの確認にも使える
画面を開かず、時間が0で得点が止まるか確かめる。

Verify the score stops at zero without opening a window.

````python
score = 0
for seconds in [2, 1, 0]:
    score = hit(score, seconds)
    print(seconds, score)
assert hit(5, 0) == 5
````
