Tkinterではボタンが関数を呼びます。下のコードを`.py`ファイルに保存し、各行の役割を説明します。

```python
import tkinter as tk
root = tk.Tk()
root.geometry('300x160')
label = tk.Label(root, text='準備完了')
label.pack()
def greet():
    label.config(text='こんにちは')
tk.Button(root, text='実行', command=greet).pack()
root.mainloop()
```
