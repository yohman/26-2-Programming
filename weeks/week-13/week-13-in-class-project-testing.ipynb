A point inside a target rectangle is a hit. Decide and check whether edges count.

```python
def inside(x, y, left, top, width, height):
    return left <= x <= left + width and top <= y <= top + height

print(inside(20, 20, 10, 10, 30, 30))
print(inside(9, 20, 10, 10, 30, 30))
assert inside(10, 10, 10, 10, 30, 30)
```
