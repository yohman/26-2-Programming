## sample.pyファイルを作成する

Week13フォルダを作成し、その中にsample.pyファイルを作成してください。

## sample.pyの中身

````python
#codimg:utf-8
#tkinterをインポートする
import tkinter as tk

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#画面を表示し続けるためのおまじないコード
root.mainloop()
````

まず、ウィンドウを作成するための`tkinter`をtkという名前でインポートします。そして、

1.   `tk.Tk()`でウインドウを作成
2.   root.geometry("400x300")で400×300というウィンドウサイズを指定
3.   root.title("タイトル")でウィンドウに表示するタイトルを指定

します。一番最後の`root.mainloop()`はウィンドウを表示し続けるためのおまじないです。

2.で指定している400×300の部分を変更するとウィンドウサイズが変わります。3.で指定している「タイトル」を変更すれば、ウィンドウ上部に表示されるタイトルが変更されます。

## ウィンドウの中に文字を表示してみる

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#テキストを入れる（ラベルを入れる）
lavel1=tk.Label(text="ここに書いた文字が表示される")
lavel1.place(x=60,y=50)


root.mainloop()
````

lavel1という変数にtkinterのLabel()メソッドを使って、表示したい文字列を指定してあげます。そうしたら次に、place()メソッドを使って、表示したい場所を$x,y$で指定します。このときの$x,y$は左上を$x=0,y=0$として下方向と右方向に正の座標を指定してあげます。**上のプログラムをsample.pyに上書きして、実行してみてください。起動しているアプリウィンドウを一度閉じると、再度`python sample.py`が実行できるようになります。**

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#テキストを入れる（ラベルを入れる）
lavel1=tk.Label(text="ここに書いた文字が表示される1")
lavel1.place(x=60,y=50)

lavel2=tk.Label(text="ここに書いた文字が表示される2")
lavel2.place(x=60,y=100)

root.mainloop()
````

ラベルを２つにすれば、文字列が２つ表示されます。このとき、placeは変更する必要があることに注意してください。

## テキストボックスを作成する

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#ラベルを作る
lavel1=tk.Label(text="ここに書いた文字が表示される")
lavel1.place(x=60,y=50)

#テキストボックスを作る
box1=tk.Entry(width=20)
box1.place(x=60,y=80)

root.mainloop()
````

今度は、テキストボックス（入力欄）を作成します。`tk.Entry()`を使うことで、テキストボックスを作成できます。()の中にwidthを指定することで、テキストボックスの長さを指定できます。今の場合、box1という変数にテキストボックスを作成しています。そして、placeを使ってテキストボックスを置く場所を指定します。label1とかぶらないようにしましょう。

## ボタンを作成する

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#ラベルを作る
lavel1=tk.Label(text="ここに書いた文字が表示される")
lavel1.place(x=60,y=50)

#テキストボックスを作る
box1=tk.Entry(width=20)
box1.place(x=60,y=80)

#ボタンを作成する
#ボタンを作る
buttonl=tk.Button(root,text="押すな危険",font=("Halvetica",14))
buttonl.place(x=140,y=130)

root.mainloop()
````

次に、ボタンを作ってみます。Buttonメソッドを使用することで、ボタンを作成できます。()の中には、どのウィンドウにボタンを置くか、ボタンになんと表示するかなどを指定します。いま、一番最初に
```
root=tk.Tk()
```
としているので、ウィンドウは`root`という名前です。なので、Button()の中に、ボタンを表示するウィンドウの名前としてrootを入れています。

　上のプログラムをsample.pyに上書きして、実行してみてください。今はボタンを押しても何も反応しません。では次に、ボタンを押したら動作する内容を追加しましょう。

## ボタンを押したら動作する内容の作成
なんでも良いのですが、まずは簡単に、入力された数値の倍を計算するような動作を設定しましょう。

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

def Buttonclick():
    a=int(box1.get())
    return a * 2

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#ラベルを作る
lavel1=tk.Label(text="ここに書いた文字が表示される")
lavel1.place(x=60,y=50)

#テキストボックスを作る
box1=tk.Entry(width=20)
box1.place(x=60,y=80)

#ボタンを作成する
#ボタンを作る
buttonl=tk.Button(root,text="押すな危険",font=("Halvetica",14),command=Buttonclick)
buttonl.place(x=140,y=130)

root.mainloop()
````

Buttonclickという関数を作成しました。Buttonclickでは、box1に入力された値の２倍を計算して返すようになっています。そして、ボタンを押したらButtonclickが実行されるようにしましょう。`command=Buttonclick`でボタンを押したときにどのような動作をするのか指定できます。上のプログラムをsample.pyに上書きして実行してみましょう。

当然ですが、計算結果を出力する場所を作っていないので、まだ何も起きません。そこで、計算結果を出力する箱を作成して、そこに計算結果を出力するようにしましょう。

````python
#codimg:utf-8
import tkinter as tk#tkinterをインポートする

def Buttonclick():
    a=int(box1.get())

    box2.delete(0,tk.END)#box2の中身を0文字目から削除
    box2.insert(0,a*2)#box2にa*2の値を0文字目から挿入

#ウィンドウを作る
root=tk.Tk()
root.geometry("400x300")#ウィンドウサイズの指定
root.title("タイトル")#ウィンドウに表示するタイトルを指定

#ラベルを作る
lavel1=tk.Label(text="ここに書いた文字が表示される")
lavel1.place(x=60,y=50)

lavel2=tk.Label(text="計算結果の出力")
lavel2.place(x=60,y=190)

#テキストボックスを作る
#数値の入力用
box1=tk.Entry(width=20)
box1.place(x=60,y=80)

#計算結果の出力用
box2=tk.Entry(width=20)
box2.place(x=60,y=220)


#ボタンを作成する
#ボタンを作る
buttonl=tk.Button(root,text="押すな危険",font=("Halvetica",14),command=Buttonclick)
buttonl.place(x=140,y=130)



root.mainloop()
````

実は、tkinterでは、作成したテキストボックスに計算結果などを表示する場合
```
box2.delete(0,tk.END)#box2の中身を0文字目から削除
box2.insert(0,a*2)#box2にa*2の値を0文字目から挿入
```
のようにinsertメソッドを使います。まずはdeleteで、box2の中身を空っぽにして、insertでa*2をテキストボックス中に出力するようにします。このとき、()の中の最初に0が入れられていますが、これはbox2の0文字目（左端の位置）からa×2の値を表示してねという意味です。
