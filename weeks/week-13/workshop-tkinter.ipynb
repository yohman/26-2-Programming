# 任意：関数をボタンから呼ぶ

Tkinterの画面が使える環境向けの発展。新しい完成条件ではありません。下のコードをbudget_window.pyへ保存してTerminalで実行できます。Notebookで動かした場合は、画面を閉じるまでCellは実行中です。

command=show_balance は関数を渡します。show_balance()と書くと今呼んでしまいます。中心のremaining関数はNotebookやTerminalと同じです。

```python
import tkinter as tk

def remaining(budget, price):
    return budget - price

def show_balance():
    try:
        amount = int(entry.get())
        result.config(text=f"Remaining: {remaining(amount, 470)} yen")
    except ValueError:
        result.config(text="Enter a whole number, such as 1000.")

root = tk.Tk()
root.title("A function behind a button")
tk.Label(root, text="Budget in yen (plan price: 470)").pack(padx=20, pady=10)
entry = tk.Entry(root)
entry.insert(0, "1000")
entry.pack()
tk.Button(root, text="Calculate", command=show_balance).pack(pady=10)
result = tk.Label(root, text="")
result.pack(pady=10)
root.mainloop()
```
