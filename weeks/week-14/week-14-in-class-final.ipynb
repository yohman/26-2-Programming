## 形式練習：修正を確かめる
本番の解答ではない。予算と同じ値を含める、いつもの境界確認。

This is practice, not an exam answer. Check an exactly affordable value.

```python
def keep_prices(prices, budget):
    kept = []
    for price in prices:
        if price <= budget:
            kept.append(price)
    return kept

print(keep_prices([150, 320, 550], 320))
```
